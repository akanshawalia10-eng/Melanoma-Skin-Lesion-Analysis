```python
# Build the PAD-UFES-20 image-only external evaluation dataset and loader

import os
from PIL import Image

import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

dataset_path = "/kaggle/input/datasets/maxjen/pad-ufes-20/PAD-UFES-20"

external_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


class PADUFES20ExternalDataset(Dataset):
    def __init__(self, dataframe, root_dir, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.root_dir = root_dir
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        image_path = os.path.join(
            self.root_dir,
            str(row["img_id"])
        )

        image = Image.open(image_path).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        label = int(row["binary_label"])

        return image, label


external_dataset = PADUFES20ExternalDataset(
    dataframe=external_metadata,
    root_dir=os.path.join(dataset_path, "Dataset"),
    transform=external_transform
)

external_loader = DataLoader(
    external_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("=" * 60)
print("PAD-UFES-20 EXTERNAL EVALUATION LOADER")
print("=" * 60)

print("Dataset size:", len(external_dataset))
print("Number of batches:", len(external_loader))
print("Batch size:", external_loader.batch_size)

images, labels = next(iter(external_loader))

print("\nFirst batch:")
print("Image tensor shape:", images.shape)
print("Image tensor dtype:", images.dtype)
print("Image tensor device:", images.device)
print("Label tensor shape:", labels.shape)
print("Label tensor dtype:", labels.dtype)

print("\nPixel range after ImageNet normalization:")
print("Minimum:", images.min().item())
print("Maximum:", images.max().item())

print("\nFirst 10 labels:")
print(labels[:10].tolist())

print("\n" + "=" * 60)
print("LOADER READY")
print("=" * 60)
```

```python
# Evaluate all five frozen HAM10000 models on the complete PAD-UFES-20 dataset

import json
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

pad_external_results = {}

for model_name, model in loaded_models.items():
    print("=" * 60)
    print(f"EVALUATING {model_name} ON PAD-UFES-20")
    print("=" * 60)

    model.eval()

    all_labels = []
    all_predictions = []
    all_probabilities = []
    total_loss = 0.0
    total_samples = 0

    criterion = torch.nn.CrossEntropyLoss()

    with torch.no_grad():
        for images, labels in pad_loader:
            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            outputs = model(images)
            loss = criterion(outputs, labels)

            probabilities = torch.softmax(outputs, dim=1)
            melanoma_probabilities = probabilities[:, 1]
            predictions = torch.argmax(outputs, dim=1)

            batch_size = labels.size(0)

            total_loss += loss.item() * batch_size
            total_samples += batch_size

            all_labels.extend(labels.cpu().numpy())
            all_predictions.extend(predictions.cpu().numpy())
            all_probabilities.extend(melanoma_probabilities.cpu().numpy())

    all_labels = np.array(all_labels)
    all_predictions = np.array(all_predictions)
    all_probabilities = np.array(all_probabilities)

    test_loss = total_loss / total_samples
    accuracy = accuracy_score(all_labels, all_predictions)
    precision = precision_score(
        all_labels,
        all_predictions,
        zero_division=0
    )
    recall = recall_score(
        all_labels,
        all_predictions,
        zero_division=0
    )
    f1 = f1_score(
        all_labels,
        all_predictions,
        zero_division=0
    )
    roc_auc = roc_auc_score(
        all_labels,
        all_probabilities
    )
    confusion = confusion_matrix(
        all_labels,
        all_predictions
    )

    pad_external_results[model_name] = {
        "model": model_name,
        "dataset": "PAD-UFES-20",
        "test_loss": float(test_loss),
        "test_accuracy": float(accuracy),
        "test_precision": float(precision),
        "test_recall": float(recall),
        "test_f1": float(f1),
        "test_roc_auc": float(roc_auc),
        "confusion_matrix": confusion.tolist(),
        "num_test_images": int(total_samples)
    }

    print(f"Loss:      {test_loss:.6f}")
    print(f"Accuracy:  {accuracy:.6f}")
    print(f"Precision: {precision:.6f}")
    print(f"Recall:    {recall:.6f}")
    print(f"F1:        {f1:.6f}")
    print(f"ROC-AUC:   {roc_auc:.6f}")
    print("Confusion matrix:")
    print(confusion)

print("=" * 60)
print("PAD-UFES-20 EXTERNAL EVALUATION COMPLETE")
print("=" * 60)
```

```python
# Implement the Lightweight Attention Melanoma Network

import torch
import torch.nn as nn


class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=8):
        super().__init__()

        hidden = max(channels // reduction, 8)

        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.mlp = nn.Sequential(
            nn.Conv2d(channels, hidden, kernel_size=1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden, channels, kernel_size=1, bias=False)
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_attention = self.mlp(self.avg_pool(x))
        max_attention = self.mlp(self.max_pool(x))

        attention = self.sigmoid(avg_attention + max_attention)

        return x * attention


class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()

        padding = kernel_size // 2

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size=kernel_size,
            padding=padding,
            bias=False
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_attention = torch.mean(x, dim=1, keepdim=True)
        max_attention, _ = torch.max(x, dim=1, keepdim=True)

        attention = torch.cat(
            [avg_attention, max_attention],
            dim=1
        )

        attention = self.sigmoid(self.conv(attention))

        return x * attention


class AttentionBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1):
        super().__init__()

        self.conv1 = nn.Conv2d(
            in_channels,
            out_channels,
            kernel_size=3,
            stride=stride,
            padding=1,
            bias=False
        )

        self.bn1 = nn.BatchNorm2d(out_channels)

        self.conv2 = nn.Conv2d(
            out_channels,
            out_channels,
            kernel_size=3,
            padding=1,
            bias=False
        )

        self.bn2 = nn.BatchNorm2d(out_channels)

        self.relu = nn.ReLU(inplace=True)

        self.channel_attention = ChannelAttention(out_channels)
        self.spatial_attention = SpatialAttention()

        if stride != 1 or in_channels != out_channels:
            self.shortcut = nn.Sequential(
                nn.Conv2d(
                    in_channels,
                    out_channels,
                    kernel_size=1,
                    stride=stride,
                    bias=False
                ),
                nn.BatchNorm2d(out_channels)
            )
        else:
            self.shortcut = nn.Identity()

    def forward(self, x):
        identity = self.shortcut(x)

        out = self.conv1(x)
        out = self.bn1(out)
        out = self.relu(out)

        out = self.conv2(out)
        out = self.bn2(out)

        out = self.channel_attention(out)
        out = self.spatial_attention(out)

        out = out + identity
        out = self.relu(out)

        return out


class LAMNet(nn.Module):
    def __init__(self, num_classes=2):
        super().__init__()

        self.stem = nn.Sequential(
            nn.Conv2d(
                3,
                32,
                kernel_size=3,
                stride=2,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )

        self.block1 = AttentionBlock(
            32,
            32,
            stride=1
        )

        self.block2 = AttentionBlock(
            32,
            64,
            stride=2
        )

        self.block3 = AttentionBlock(
            64,
            128,
            stride=2
        )

        self.block4 = AttentionBlock(
            128,
            256,
            stride=2
        )

        self.block5 = AttentionBlock(
            256,
            256,
            stride=1
        )

        self.global_pool = nn.AdaptiveAvgPool2d(1)

        self.dropout = nn.Dropout(p=0.30)

        self.classifier = nn.Linear(
            256,
            num_classes
        )

    def forward(self, x):
        x = self.stem(x)

        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        x = self.block5(x)

        x = self.global_pool(x)

        x = torch.flatten(x, 1)

        x = self.dropout(x)

        x = self.classifier(x)

        return x


print("=" * 60)
print("LAMNet ARCHITECTURE CREATED")
print("=" * 60)

model = LAMNet(num_classes=2)

print(model)
```

```python
# Recreate the verified LAMNet architecture

import torch
import torch.nn as nn


class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()

        hidden_channels = max(channels // reduction, 1)

        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.mlp = nn.Sequential(
            nn.Conv2d(channels, hidden_channels, kernel_size=1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden_channels, channels, kernel_size=1, bias=False)
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_attention = self.mlp(self.avg_pool(x))
        max_attention = self.mlp(self.max_pool(x))

        attention = avg_attention + max_attention

        return x * self.sigmoid(attention)


class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()

        padding = kernel_size // 2

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size=kernel_size,
            padding=padding,
            bias=False
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_attention = torch.mean(x, dim=1, keepdim=True)
        max_attention, _ = torch.max(x, dim=1, keepdim=True)

        attention = torch.cat(
            [avg_attention, max_attention],
            dim=1
        )

        attention = self.sigmoid(self.conv(attention))

        return x * attention


class AttentionBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1):
        super().__init__()

        self.conv1 = nn.Conv2d(
            in_channels,
            out_channels,
            kernel_size=3,
            stride=stride,
            padding=1,
            bias=False
        )

        self.bn1 = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)

        self.conv2 = nn.Conv2d(
            out_channels,
            out_channels,
            kernel_size=3,
            padding=1,
            bias=False
        )

        self.bn2 = nn.BatchNorm2d(out_channels)

        self.channel_attention = ChannelAttention(out_channels)
        self.spatial_attention = SpatialAttention()

        if stride != 1 or in_channels != out_channels:
            self.shortcut = nn.Sequential(
                nn.Conv2d(
                    in_channels,
                    out_channels,
                    kernel_size=1,
                    stride=stride,
                    bias=False
                ),
                nn.BatchNorm2d(out_channels)
            )
        else:
            self.shortcut = nn.Identity()

    def forward(self, x):
        identity = self.shortcut(x)

        out = self.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))

        out = self.channel_attention(out)
        out = self.spatial_attention(out)

        out = out + identity
        out = self.relu(out)

        return out


class LAMNet(nn.Module):
    def __init__(self, num_classes=2):
        super().__init__()

        self.stem = nn.Sequential(
            nn.Conv2d(
                3,
                32,
                kernel_size=3,
                stride=2,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )

        self.block1 = AttentionBlock(32, 32)
        self.block2 = AttentionBlock(32, 64, stride=2)
        self.block3 = AttentionBlock(64, 128, stride=2)
        self.block4 = AttentionBlock(128, 256, stride=2)
        self.block5 = AttentionBlock(256, 256)

        self.global_pool = nn.AdaptiveAvgPool2d(1)

        self.dropout = nn.Dropout(0.30)

        self.classifier = nn.Linear(256, num_classes)

    def forward(self, x):
        x = self.stem(x)

        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        x = self.block5(x)

        x = self.global_pool(x)
        x = torch.flatten(x, 1)

        x = self.dropout(x)

        return self.classifier(x)


print("LAMNet architecture restored")
```

```python
# Restore HAM10000 classification datasets for LAMNet

import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset
from torchvision import transforms

HAM_ROOT = (
    "/kaggle/input/datasets/akansha1009/"
    "ham10000-preprocessed-segmentation"
)

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]


class HAM10000ClassificationDataset(Dataset):
    def __init__(self, split, transform=None):
        self.split = split
        self.transform = transform

        split_path = f"{HAM_ROOT}/{split}"

        self.images = np.load(
            f"{split_path}/images.npy",
            mmap_mode="r"
        )

        self.metadata = pd.read_csv(
            f"{split_path}/metadata.csv"
        )

        self.labels = (
            self.metadata["dx"] == "mel"
        ).astype(int).values

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, index):
        image = self.images[index]
        label = int(self.labels[index])

        image = torch.from_numpy(
            np.asarray(image)
        ).permute(2, 0, 1).float()

        if self.transform is not None:
            image = self.transform(image)

        return image, label


train_transform = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.2),
    transforms.RandomRotation(15),
    transforms.ColorJitter(
        brightness=0.15,
        contrast=0.15,
        saturation=0.10,
        hue=0.02
    ),
    transforms.Normalize(
        mean=IMAGENET_MEAN,
        std=IMAGENET_STD
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.Normalize(
        mean=IMAGENET_MEAN,
        std=IMAGENET_STD
    )
])

train_dataset = HAM10000ClassificationDataset(
    "train",
    transform=train_transform
)

val_dataset = HAM10000ClassificationDataset(
    "validation",
    transform=eval_transform
)

test_dataset = HAM10000ClassificationDataset(
    "test",
    transform=eval_transform
)

print("=" * 60)
print("HAM10000 DATASETS RESTORED")
print("=" * 60)
print(f"Training samples: {len(train_dataset):,}")
print(f"Validation samples: {len(val_dataset):,}")
print(f"Test samples: {len(test_dataset):,}")
```

```python
# Purpose: Prepare the five trained benchmark model architectures and checkpoint paths for fairness inference

import torch
import torch.nn as nn
from torchvision import models

device = torch.device(
    "cuda:0" if torch.cuda.is_available() else "cpu"
)

checkpoint_paths = {
    "AlexNet": Path(
        "/kaggle/input/datasets/akansha1009/5-trained-models/"
        "HAM10000_AlexNet_Balanced_Baseline_Backup/"
        "ham10000_alexnet_balanced_best.pth"
    ),
    "VGG16": Path(
        "/kaggle/input/datasets/akansha1009/5-trained-models/"
        "HAM10000_VGG16_Balanced_Baseline_Backup/"
        "ham10000_vgg16_balanced_best.pth"
    ),
    "VGG19": Path(
        "/kaggle/input/datasets/akansha1009/5-trained-models/"
        "HAM10000_VGG19_Balanced_Baseline_Backup/"
        "ham10000_vgg19_balanced_best.pth"
    ),
    "ResNet50": Path(
        "/kaggle/input/datasets/akansha1009/5-trained-models/"
        "HAM10000_ResNet50_Balanced_Baseline_Backup/"
        "ham10000_resnet50_balanced_best.pth"
    ),
    "EfficientNet-B0": Path(
        "/kaggle/input/datasets/akansha1009/5-trained-models/"
        "HAM10000_EfficientNetB0_Balanced_Baseline_Backup/"
        "ham10000_efficientnet_b0_balanced_best.pth"
    )
}

def build_model(model_name):
    if model_name == "AlexNet":
        model = models.alexnet(weights=None)
        model.classifier[6] = nn.Linear(
            model.classifier[6].in_features,
            2
        )

    elif model_name == "VGG16":
        model = models.vgg16(weights=None)
        model.classifier[6] = nn.Linear(
            model.classifier[6].in_features,
            2
        )

    elif model_name == "VGG19":
        model = models.vgg19(weights=None)
        model.classifier[6] = nn.Linear(
            model.classifier[6].in_features,
            2
        )

    elif model_name == "ResNet50":
        model = models.resnet50(weights=None)
        model.fc = nn.Linear(
            model.fc.in_features,
            2
        )

    elif model_name == "EfficientNet-B0":
        model = models.efficientnet_b0(weights=None)
        model.classifier[1] = nn.Linear(
            model.classifier[1].in_features,
            2
        )

    else:
        raise ValueError(f"Unknown model: {model_name}")

    return model


print("=" * 70)
print("FAIRNESS MODEL PREPARATION")
print("=" * 70)

print(f"\nDevice: {device}")

for model_name, checkpoint_path in checkpoint_paths.items():
    print(
        f"{model_name}: "
        f"{'FOUND' if checkpoint_path.exists() else 'MISSING'}"
    )

print("\nModel architectures prepared:")
for model_name in checkpoint_paths:
    model = build_model(model_name)
    print(
        f"{model_name}: "
        f"{sum(p.numel() for p in model.parameters()):,} parameters"
    )
    del model

print("=" * 70)
print("FAIRNESS MODEL PREPARATION COMPLETE")
print("=" * 70)
```

```python
# Purpose: Restore the standardized PAD-UFES-20 evaluation DataLoader with the required transforms import

from torchvision import transforms
from torch.utils.data import Dataset, DataLoader
from PIL import Image

pad_eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class PADUFES20FairnessDataset(Dataset):
    def __init__(self, dataframe, image_root, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.image_root = Path(image_root)
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        image_path = self.image_root / row["img_id"]

        image = Image.open(image_path).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        label = int(row["diagnostic"] == "MEL")

        return image, torch.tensor(label, dtype=torch.long)


pad_dataset = PADUFES20FairnessDataset(
    dataframe=fairness_df,
    image_root=PAD_IMAGE_ROOT,
    transform=pad_eval_transform
)

pad_loader = DataLoader(
    pad_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("=" * 70)
print("PAD-UFES-20 FAIRNESS DATALOADER RESTORED")
print("=" * 70)

print(f"\nImage root: {PAD_IMAGE_ROOT}")
print(f"Dataset samples: {len(pad_dataset)}")
print(f"DataLoader batches: {len(pad_loader)}")
print(f"Batch size: {pad_loader.batch_size}")
print(f"Shuffle: False")
print(f"Workers: {pad_loader.num_workers}")

print(f"\nExpected samples: 2298")
print(f"Sample count correct: {len(pad_dataset) == 2298}")

print("=" * 70)
print("PAD-UFES-20 FAIRNESS DATALOADER RESTORATION COMPLETE")
print("=" * 70)
```

```python
# Purpose: Calculate ResNet50 performance separately for Fitzpatrick I-IV

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

fairness_groups_for_analysis = [
    "Fitzpatrick I",
    "Fitzpatrick II",
    "Fitzpatrick III",
    "Fitzpatrick IV"
]

subgroup_results = []

for group in fairness_groups_for_analysis:

    group_df = prediction_df[
        prediction_df["fairness_group"] == group
    ].copy()

    y_true = group_df["true_label"].to_numpy()
    y_pred = group_df["predicted_label"].to_numpy()
    y_prob = group_df["melanoma_probability"].to_numpy()

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    ).ravel()

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    if len(np.unique(y_true)) == 2:
        auc = roc_auc_score(
            y_true,
            y_prob
        )
    else:
        auc = np.nan

    subgroup_results.append({
        "Fitzpatrick Group": group,
        "Samples": len(group_df),
        "Melanoma": int(y_true.sum()),
        "Non-Melanoma": int((y_true == 0).sum()),
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(
            y_true,
            y_pred,
            zero_division=0
        ),
        "Recall": recall_score(
            y_true,
            y_pred,
            zero_division=0
        ),
        "Specificity": specificity,
        "F1": f1_score(
            y_true,
            y_pred,
            zero_division=0
        ),
        "ROC-AUC": auc,
        "TN": int(tn),
        "FP": int(fp),
        "FN": int(fn),
        "TP": int(tp)
    })

resnet50_fairness_results = pd.DataFrame(
    subgroup_results
)

print("=" * 70)
print("RESNET50 — FITZPATRICK SUBGROUP PERFORMANCE")
print("=" * 70)

print(
    resnet50_fairness_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\n" + "=" * 70)
print("RESNET50 SUBGROUP METRICS COMPLETE")
print("=" * 70)
```

```python
# Purpose: Estimate 95 percent bootstrap confidence intervals for subgroup melanoma recall

from sklearn.utils import resample

np.random.seed(42)

bootstrap_iterations = 2000

bootstrap_results = []

for group in fairness_groups_for_analysis:

    group_df = prediction_df[
        prediction_df["fairness_group"] == group
    ].copy()

    y_true = group_df["true_label"].to_numpy()
    y_pred = group_df["predicted_label"].to_numpy()

    recall_samples = []

    for _ in range(bootstrap_iterations):

        indices = np.random.randint(
            0,
            len(y_true),
            size=len(y_true)
        )

        y_true_boot = y_true[indices]
        y_pred_boot = y_pred[indices]

        recall_value = recall_score(
            y_true_boot,
            y_pred_boot,
            zero_division=0
        )

        recall_samples.append(recall_value)

    recall_samples = np.asarray(
        recall_samples
    )

    lower = np.percentile(
        recall_samples,
        2.5
    )

    upper = np.percentile(
        recall_samples,
        97.5
    )

    bootstrap_results.append({
        "Fitzpatrick Group": group,
        "Samples": len(group_df),
        "Melanoma": int(y_true.sum()),
        "Observed Recall": recall_score(
            y_true,
            y_pred,
            zero_division=0
        ),
        "Bootstrap 95% CI Lower": lower,
        "Bootstrap 95% CI Upper": upper
    })

resnet50_recall_ci = pd.DataFrame(
    bootstrap_results
)

print("=" * 70)
print("RESNET50 — BOOTSTRAP RECALL UNCERTAINTY")
print("=" * 70)

print(
    resnet50_recall_ci.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nBootstrap iterations:", bootstrap_iterations)

print("=" * 70)
print("BOOTSTRAP RECALL ANALYSIS COMPLETE")
print("=" * 70)
```

```python
# Purpose: Estimate 95 percent bootstrap confidence intervals for subgroup specificity

np.random.seed(42)

bootstrap_iterations = 2000

specificity_results = []

for group in fairness_groups_for_analysis:

    group_df = prediction_df[
        prediction_df["fairness_group"] == group
    ].copy()

    y_true = group_df["true_label"].to_numpy()
    y_pred = group_df["predicted_label"].to_numpy()

    specificity_samples = []

    for _ in range(bootstrap_iterations):

        indices = np.random.randint(
            0,
            len(y_true),
            size=len(y_true)
        )

        y_true_boot = y_true[indices]
        y_pred_boot = y_pred[indices]

        tn, fp, fn, tp = confusion_matrix(
            y_true_boot,
            y_pred_boot,
            labels=[0, 1]
        ).ravel()

        specificity_value = (
            tn / (tn + fp)
            if (tn + fp) > 0
            else np.nan
        )

        if not np.isnan(specificity_value):
            specificity_samples.append(
                specificity_value
            )

    specificity_samples = np.asarray(
        specificity_samples
    )

    lower = np.percentile(
        specificity_samples,
        2.5
    )

    upper = np.percentile(
        specificity_samples,
        97.5
    )

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    ).ravel()

    observed_specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    specificity_results.append({
        "Fitzpatrick Group": group,
        "Samples": len(group_df),
        "Non-Melanoma": int((y_true == 0).sum()),
        "Observed Specificity": observed_specificity,
        "Bootstrap 95% CI Lower": lower,
        "Bootstrap 95% CI Upper": upper
    })

resnet50_specificity_ci = pd.DataFrame(
    specificity_results
)

print("=" * 70)
print("RESNET50 — BOOTSTRAP SPECIFICITY UNCERTAINTY")
print("=" * 70)

print(
    resnet50_specificity_ci.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nBootstrap iterations:", bootstrap_iterations)

print("=" * 70)
print("BOOTSTRAP SPECIFICITY ANALYSIS COMPLETE")
print("=" * 70)
```

```python
# Purpose: Calculate formal worst-group fairness gaps from the final Fitzpatrick subgroup results

fairness_metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "Specificity",
    "F1",
    "ROC-AUC"
]

fairness_gap_results = []

for metric in fairness_metrics:

    metric_values = fairness_summary[
        ["Fitzpatrick Group", metric]
    ].dropna()

    minimum_row = metric_values.loc[
        metric_values[metric].idxmin()
    ]

    maximum_row = metric_values.loc[
        metric_values[metric].idxmax()
    ]

    gap = maximum_row[metric] - minimum_row[metric]

    fairness_gap_results.append({
        "Metric": metric,
        "Best Group": maximum_row["Fitzpatrick Group"],
        "Best Value": maximum_row[metric],
        "Worst Group": minimum_row["Fitzpatrick Group"],
        "Worst Value": minimum_row[metric],
        "Absolute Gap": gap,
        "Gap Percentage Points": gap * 100
    })

fairness_gap_summary = pd.DataFrame(fairness_gap_results)

print("=" * 100)
print("OBJECTIVE 4 — FORMAL FAIRNESS GAP SUMMARY")
print("=" * 100)

print(
    fairness_gap_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\n" + "=" * 100)
print("FAIRNESS GAP CALCULATION COMPLETE")
print("=" * 100)
```

```python
# Purpose: Restore the exact LAMNet architecture used for the official HAM10000 baseline

class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()

        hidden_channels = max(channels // reduction, 1)

        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.mlp = nn.Sequential(
            nn.Conv2d(channels, hidden_channels, kernel_size=1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden_channels, channels, kernel_size=1, bias=False)
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_attention = self.mlp(self.avg_pool(x))
        max_attention = self.mlp(self.max_pool(x))

        attention = avg_attention + max_attention

        return self.sigmoid(attention)


class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()

        padding = kernel_size // 2

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size=kernel_size,
            padding=padding,
            bias=False
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        mean_attention = torch.mean(x, dim=1, keepdim=True)
        max_attention, _ = torch.max(x, dim=1, keepdim=True)

        attention = torch.cat(
            [mean_attention, max_attention],
            dim=1
        )

        return self.sigmoid(self.conv(attention))


class AttentionBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1):
        super().__init__()

        self.conv1 = nn.Conv2d(
            in_channels,
            out_channels,
            kernel_size=3,
            stride=stride,
            padding=1,
            bias=False
        )

        self.bn1 = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)

        self.conv2 = nn.Conv2d(
            out_channels,
            out_channels,
            kernel_size=3,
            stride=1,
            padding=1,
            bias=False
        )

        self.bn2 = nn.BatchNorm2d(out_channels)

        self.channel_attention = ChannelAttention(out_channels)
        self.spatial_attention = SpatialAttention()

        if stride != 1 or in_channels != out_channels:
            self.shortcut = nn.Sequential(
                nn.Conv2d(
                    in_channels,
                    out_channels,
                    kernel_size=1,
                    stride=stride,
                    bias=False
                ),
                nn.BatchNorm2d(out_channels)
            )
        else:
            self.shortcut = nn.Identity()

    def forward(self, x):
        identity = self.shortcut(x)

        out = self.conv1(x)
        out = self.bn1(out)
        out = self.relu(out)

        out = self.conv2(out)
        out = self.bn2(out)

        channel_map = self.channel_attention(out)
        out = out * channel_map

        spatial_map = self.spatial_attention(out)
        out = out * spatial_map

        out = out + identity
        out = self.relu(out)

        return out


class LAMNet(nn.Module):
    def __init__(self, num_classes=2):
        super().__init__()

        self.stem = nn.Sequential(
            nn.Conv2d(
                3,
                32,
                kernel_size=3,
                stride=2,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )

        self.block1 = AttentionBlock(32, 32, stride=1)
        self.block2 = AttentionBlock(32, 64, stride=2)
        self.block3 = AttentionBlock(64, 128, stride=2)
        self.block4 = AttentionBlock(128, 256, stride=2)
        self.block5 = AttentionBlock(256, 256, stride=1)

        self.global_pool = nn.AdaptiveAvgPool2d(1)
        self.dropout = nn.Dropout(0.30)
        self.classifier = nn.Linear(256, num_classes)

    def forward(self, x):
        x = self.stem(x)

        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        x = self.block5(x)

        x = self.global_pool(x)
        x = torch.flatten(x, 1)

        x = self.dropout(x)

        return self.classifier(x)


lamnet_test_model = LAMNet(num_classes=2)

lamnet_parameter_count = sum(
    parameter.numel()
    for parameter in lamnet_test_model.parameters()
)

print("=" * 90)
print("LAMNET ARCHITECTURE RESTORED")
print("=" * 90)

print("\nParameter count:", f"{lamnet_parameter_count:,}")
print("Expected:", "2,427,148")
print(
    "Architecture verified:",
    lamnet_parameter_count == 2_427_148
)

print("=" * 90)
```

```python
# Purpose: Create the standardized HAM10000 test dataset and deterministic evaluation loader for Objective 5

class HAM10000UncertaintyDataset(Dataset):
    def __init__(self, images, labels, transform=None):
        self.images = images
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, index):
        image = np.asarray(
            self.images[index],
            dtype=np.float32
        )

        image = np.clip(
            image,
            0.0,
            1.0
        )

        image = (
            image * 255
        ).astype(np.uint8)

        if self.transform is not None:
            image = self.transform(image)
        else:
            image = torch.from_numpy(
                image.transpose(2, 0, 1)
            ).float() / 255.0

        label = torch.tensor(
            self.labels[index],
            dtype=torch.long
        )

        return image, label


HAM_TEST_EVAL_TRANSFORM = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


ham_test_dataset = HAM10000UncertaintyDataset(
    images=ham_test_images,
    labels=ham_test_labels,
    transform=HAM_TEST_EVAL_TRANSFORM
)


ham_test_loader = DataLoader(
    ham_test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


print("=" * 90)
print("OBJECTIVE 5 — HAM10000 UNCERTAINTY TEST LOADER")
print("=" * 90)

print("\nDataset samples:", len(ham_test_dataset))
print("DataLoader batches:", len(ham_test_loader))
print("Batch size:", ham_test_loader.batch_size)
print("Shuffle:", False)

print("\nEvaluation transform:")
print(HAM_TEST_EVAL_TRANSFORM)

print("\nExpected samples: 1,502")
print(
    "Dataset size verified:",
    len(ham_test_dataset) == 1502
)

print(
    "Batch count verified:",
    len(ham_test_loader) == 47
)

print("=" * 90)
print("HAM10000 UNCERTAINTY TEST LOADER READY")
print("=" * 90)
```

```python
# Purpose: Verify the exact five benchmark architectures used for Objective 5 probability inference

def build_benchmark_model(model_name, num_classes=2):
    if model_name == "AlexNet":
        model = torchvision.models.alexnet(weights=None)
        model.classifier[6] = nn.Linear(
            model.classifier[6].in_features,
            num_classes
        )

    elif model_name == "VGG16":
        model = torchvision.models.vgg16(weights=None)
        model.classifier[6] = nn.Linear(
            model.classifier[6].in_features,
            num_classes
        )

    elif model_name == "VGG19":
        model = torchvision.models.vgg19(weights=None)
        model.classifier[6] = nn.Linear(
            model.classifier[6].in_features,
            num_classes
        )

    elif model_name == "ResNet50":
        model = torchvision.models.resnet50(weights=None)
        model.fc = nn.Linear(
            model.fc.in_features,
            num_classes
        )

    elif model_name == "EfficientNet-B0":
        model = torchvision.models.efficientnet_b0(weights=None)
        model.classifier[1] = nn.Linear(
            model.classifier[1].in_features,
            num_classes
        )

    else:
        raise ValueError(
            f"Unsupported benchmark model: {model_name}"
        )

    return model


print("=" * 90)
print("OBJECTIVE 5 — BENCHMARK MODEL ARCHITECTURE VERIFICATION")
print("=" * 90)

for model_name in BENCHMARK_CHECKPOINT_PATHS:
    test_model = build_benchmark_model(model_name)

    parameter_count = sum(
        parameter.numel()
        for parameter in test_model.parameters()
    )

    print(
        f"{model_name:18s} | "
        f"Parameters: {parameter_count:,}"
    )

    del test_model
    gc.collect()

print("=" * 90)
print("BENCHMARK MODEL ARCHITECTURES VERIFIED")
print("=" * 90)
```

```python
# Purpose: Define a reproducible function for generating and saving individual test-set probabilities

OBJECTIVE5_DIR = Path("/kaggle/working/Objective5_Uncertainty")
OBJECTIVE5_DIR.mkdir(
    parents=True,
    exist_ok=True
)


def load_checkpoint_model(model_name):
    model = build_benchmark_model(
        model_name,
        num_classes=2
    )

    checkpoint_path = BENCHMARK_CHECKPOINT_PATHS[model_name]

    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False
    )

    state_dict = checkpoint["model_state_dict"]

    if any(
        key.startswith("module.")
        for key in state_dict.keys()
    ):
        state_dict = {
            key.replace("module.", "", 1): value
            for key, value in state_dict.items()
        }

    load_result = model.load_state_dict(
        state_dict
    )

    if (
        len(load_result.missing_keys) > 0
        or len(load_result.unexpected_keys) > 0
    ):
        raise RuntimeError(
            f"Checkpoint mismatch for {model_name}: "
            f"missing={load_result.missing_keys}, "
            f"unexpected={load_result.unexpected_keys}"
        )

    model = model.to(device)
    model.eval()

    return model, checkpoint


def generate_model_probabilities(
    model_name,
    model,
    loader,
    output_path
):
    all_probabilities = []
    all_predictions = []
    all_labels = []

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(
                device,
                non_blocking=True
            )

            logits = model(images)

            probabilities = torch.softmax(
                logits,
                dim=1
            )

            predictions = torch.argmax(
                probabilities,
                dim=1
            )

            all_probabilities.append(
                probabilities[:, 1]
                .detach()
                .cpu()
                .numpy()
            )

            all_predictions.append(
                predictions
                .detach()
                .cpu()
                .numpy()
            )

            all_labels.append(
                labels.numpy()
            )

    melanoma_probabilities = np.concatenate(
        all_probabilities
    )

    predictions = np.concatenate(
        all_predictions
    )

    labels = np.concatenate(
        all_labels
    )

    if len(labels) != 1502:
        raise RuntimeError(
            f"Expected 1502 predictions, "
            f"got {len(labels)}"
        )

    probability_df = pd.DataFrame({
        "sample_index": np.arange(len(labels)),
        "true_label": labels,
        "melanoma_probability": melanoma_probabilities,
        "predicted_label": predictions
    })

    probability_df.to_csv(
        output_path,
        index=False
    )

    return probability_df


print("=" * 90)
print("OBJECTIVE 5 — PROBABILITY INFERENCE PIPELINE READY")
print("=" * 90)

print("\nOutput directory:")
print(OBJECTIVE5_DIR)

print("\nFunction 1:")
print("load_checkpoint_model()")

print("\nFunction 2:")
print("generate_model_probabilities()")

print("\nExpected models: 6")
print("Expected test samples per model: 1,502")

print("=" * 90)
print("PROBABILITY INFERENCE PIPELINE READY")
print("=" * 90)
```

```python
# Purpose: Verify that all six model probability artifacts align to the same HAM10000 test samples and labels

print("=" * 90)
print("OBJECTIVE 5 — SIX-MODEL PROBABILITY ALIGNMENT AUDIT")
print("=" * 90)

probability_files = {
    "AlexNet": OBJECTIVE5_DIR / "AlexNet_HAM10000_Test_Probabilities.csv",
    "VGG16": OBJECTIVE5_DIR / "VGG16_HAM10000_Test_Probabilities.csv",
    "VGG19": OBJECTIVE5_DIR / "VGG19_HAM10000_Test_Probabilities.csv",
    "ResNet50": OBJECTIVE5_DIR / "ResNet50_HAM10000_Test_Probabilities.csv",
    "EfficientNet-B0": OBJECTIVE5_DIR / "EfficientNetB0_HAM10000_Test_Probabilities.csv",
    "LAMNet": OBJECTIVE5_DIR / "LAMNet_HAM10000_Test_Probabilities.csv"
}

print("\nChecking probability artifacts...")

probability_data = {}

for model_name, file_path in probability_files.items():
    if not file_path.exists():
        raise FileNotFoundError(
            f"Missing probability artifact for {model_name}: {file_path}"
        )

    df = pd.read_csv(file_path)

    required_columns = {
        "sample_index",
        "true_label",
        "melanoma_probability",
        "predicted_label"
    }

    missing_columns = required_columns - set(df.columns)

    if missing_columns:
        raise RuntimeError(
            f"{model_name} is missing columns: {sorted(missing_columns)}"
        )

    probability_data[model_name] = df

    print(
        f"{model_name:<18} | "
        f"Rows: {len(df):4d} | "
        f"Columns verified: YES"
    )

print("\nVerifying row counts...")

row_counts = {
    model_name: len(df)
    for model_name, df in probability_data.items()
}

if len(set(row_counts.values())) != 1:
    raise RuntimeError(
        f"Row-count mismatch detected: {row_counts}"
    )

if list(row_counts.values())[0] != 1502:
    raise RuntimeError(
        f"Expected 1,502 rows per model, found {row_counts}"
    )

print("All six models contain exactly 1,502 rows.")

print("\nVerifying sample-index alignment...")

reference_model = "AlexNet"
reference_df = probability_data[reference_model]

for model_name, df in probability_data.items():
    if not np.array_equal(
        reference_df["sample_index"].to_numpy(),
        df["sample_index"].to_numpy()
    ):
        raise RuntimeError(
            f"Sample-index mismatch detected: {model_name}"
        )

print("Sample-index alignment: PASSED")

print("\nVerifying true-label alignment...")

for model_name, df in probability_data.items():
    if not np.array_equal(
        reference_df["true_label"].to_numpy(),
        df["true_label"].to_numpy()
    ):
        raise RuntimeError(
            f"True-label mismatch detected: {model_name}"
        )

print("True-label alignment: PASSED")

print("\nVerifying expected class distribution...")

true_labels = reference_df["true_label"].to_numpy()

melanoma_count = int(true_labels.sum())
non_melanoma_count = int((true_labels == 0).sum())

if melanoma_count != 167 or non_melanoma_count != 1335:
    raise RuntimeError(
        f"Unexpected class distribution: "
        f"melanoma={melanoma_count}, "
        f"non-melanoma={non_melanoma_count}"
    )

print(f"Melanoma samples: {melanoma_count}")
print(f"Non-melanoma samples: {non_melanoma_count}")

print("\nVerifying probability validity...")

for model_name, df in probability_data.items():
    probabilities = df["melanoma_probability"].to_numpy()

    if not np.all(np.isfinite(probabilities)):
        raise RuntimeError(
            f"Non-finite probability detected in {model_name}"
        )

    if np.any(probabilities < 0.0) or np.any(probabilities > 1.0):
        raise RuntimeError(
            f"Invalid probability detected in {model_name}"
        )

print("All probabilities are finite and within [0, 1].")

print("\nVerifying predicted-label consistency...")

for model_name, df in probability_data.items():
    probabilities = df["melanoma_probability"].to_numpy()
    stored_predictions = df["predicted_label"].to_numpy()

    reconstructed_predictions = (probabilities >= 0.5).astype(np.int64)

    if not np.array_equal(
        stored_predictions,
        reconstructed_predictions
    ):
        raise RuntimeError(
            f"Prediction mismatch detected in {model_name}"
        )

print("Predicted labels are consistent with the 0.5 threshold.")

print("\n" + "=" * 90)
print("SIX-MODEL PROBABILITY ALIGNMENT AUDIT PASSED")
print("=" * 90)

print("\nVerified:")
print("✓ Six model probability files present")
print("✓ 1,502 samples per model")
print("✓ Same sample indices")
print("✓ Same true labels")
print("✓ Same class distribution")
print("✓ All probabilities finite")
print("✓ All probabilities within [0, 1]")
print("✓ Stored predictions match the 0.5 threshold")

print("\nObjective 5 probability artifacts are ready for calibration analysis.")
```

```python
# Purpose: Calculate research-grade calibration metrics for all six models on the untouched HAM10000 test set

from sklearn.metrics import brier_score_loss
import numpy as np
import pandas as pd

print("=" * 90)
print("OBJECTIVE 5 — CALIBRATION METRICS")
print("=" * 90)

def calculate_ece(y_true, probabilities, n_bins=10):
    bin_edges = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0
    bin_results = []

    for bin_index in range(n_bins):
        lower = bin_edges[bin_index]
        upper = bin_edges[bin_index + 1]

        if bin_index == n_bins - 1:
            mask = (
                (probabilities >= lower) &
                (probabilities <= upper)
            )
        else:
            mask = (
                (probabilities >= lower) &
                (probabilities < upper)
            )

        count = int(mask.sum())

        if count == 0:
            bin_results.append({
                "bin": bin_index + 1,
                "lower": lower,
                "upper": upper,
                "count": 0,
                "mean_confidence": np.nan,
                "empirical_accuracy": np.nan,
                "absolute_gap": np.nan
            })
            continue

        mean_probability = float(probabilities[mask].mean())
        empirical_accuracy = float(y_true[mask].mean())

        absolute_gap = abs(
            mean_probability - empirical_accuracy
        )

        ece += (
            count / len(y_true)
        ) * absolute_gap

        bin_results.append({
            "bin": bin_index + 1,
            "lower": lower,
            "upper": upper,
            "count": count,
            "mean_confidence": mean_probability,
            "empirical_accuracy": empirical_accuracy,
            "absolute_gap": absolute_gap
        })

    return float(ece), pd.DataFrame(bin_results)


def calculate_mce(y_true, probabilities, n_bins=10):
    _, bin_df = calculate_ece(
        y_true,
        probabilities,
        n_bins=n_bins
    )

    valid_bins = bin_df.dropna(subset=["absolute_gap"])

    if len(valid_bins) == 0:
        return np.nan

    return float(valid_bins["absolute_gap"].max())


calibration_results = []
calibration_bins = {}

for model_name, df in probability_data.items():

    y_true = df["true_label"].to_numpy(dtype=np.int64)
    probabilities = df["melanoma_probability"].to_numpy(
        dtype=np.float64
    )

    predictions = df["predicted_label"].to_numpy(
        dtype=np.int64
    )

    brier = brier_score_loss(
        y_true,
        probabilities
    )

    ece, bin_df = calculate_ece(
        y_true,
        probabilities,
        n_bins=10
    )

    mce = calculate_mce(
        y_true,
        probabilities,
        n_bins=10
    )

    accuracy = float(
        (predictions == y_true).mean()
    )

    calibration_results.append({
        "Model": model_name,
        "Brier Score": brier,
        "ECE": ece,
        "MCE": mce,
        "Accuracy": accuracy
    })

    calibration_bins[model_name] = bin_df

calibration_summary = pd.DataFrame(
    calibration_results
).sort_values(
    by="Brier Score",
    ascending=True
).reset_index(drop=True)

print("\nCalibration results:")
print(
    calibration_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

calibration_output_path = (
    OBJECTIVE5_DIR /
    "Six_Model_Calibration_Summary.csv"
)

calibration_summary.to_csv(
    calibration_output_path,
    index=False
)

print("\nSaved artifact:")
print(calibration_output_path)

print("\n" + "=" * 90)
print("CALIBRATION METRICS COMPLETE")
print("=" * 90)
```

```python
# Purpose: Quantify prediction confidence, correctness, and overconfident errors for all six models

import numpy as np
import pandas as pd

print("=" * 90)
print("OBJECTIVE 5 — CONFIDENCE VS CORRECTNESS ANALYSIS")
print("=" * 90)

confidence_results = []
confidence_records = {}

for model_name, df in probability_data.items():

    y_true = df["true_label"].to_numpy(dtype=np.int64)
    melanoma_probability = df["melanoma_probability"].to_numpy(
        dtype=np.float64
    )
    predictions = df["predicted_label"].to_numpy(dtype=np.int64)

    confidence = np.where(
        predictions == 1,
        melanoma_probability,
        1.0 - melanoma_probability
    )

    correct = predictions == y_true

    incorrect = ~correct

    high_conf_80 = confidence >= 0.80
    high_conf_90 = confidence >= 0.90

    high_conf_errors_80 = high_conf_80 & incorrect
    high_conf_errors_90 = high_conf_90 & incorrect

    correct_confidence = confidence[correct]
    incorrect_confidence = confidence[incorrect]

    result = {
        "Model": model_name,
        "Mean Confidence": confidence.mean(),
        "Mean Correct Confidence": (
            correct_confidence.mean()
            if len(correct_confidence) > 0
            else np.nan
        ),
        "Mean Incorrect Confidence": (
            incorrect_confidence.mean()
            if len(incorrect_confidence) > 0
            else np.nan
        ),
        "Incorrect Predictions": int(incorrect.sum()),
        "High-Confidence Predictions >= 0.80": int(
            high_conf_80.sum()
        ),
        "High-Confidence Errors >= 0.80": int(
            high_conf_errors_80.sum()
        ),
        "High-Confidence Error Rate >= 0.80": (
            high_conf_errors_80.sum() /
            high_conf_80.sum()
            if high_conf_80.sum() > 0
            else np.nan
        ),
        "Very-High-Confidence Predictions >= 0.90": int(
            high_conf_90.sum()
        ),
        "Very-High-Confidence Errors >= 0.90": int(
            high_conf_errors_90.sum()
        ),
        "Very-High-Confidence Error Rate >= 0.90": (
            high_conf_errors_90.sum() /
            high_conf_90.sum()
            if high_conf_90.sum() > 0
            else np.nan
        )
    }

    confidence_results.append(result)

    confidence_records[model_name] = pd.DataFrame({
        "sample_index": df["sample_index"].to_numpy(),
        "true_label": y_true,
        "predicted_label": predictions,
        "melanoma_probability": melanoma_probability,
        "confidence": confidence,
        "correct": correct
    })

confidence_summary = pd.DataFrame(
    confidence_results
)

print("\nConfidence and correctness summary:")

print(
    confidence_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

confidence_output_path = (
    OBJECTIVE5_DIR /
    "Six_Model_Confidence_Correctness_Summary.csv"
)

confidence_summary.to_csv(
    confidence_output_path,
    index=False
)

print("\nSaved summary:")
print(confidence_output_path)

print("\n" + "=" * 90)
print("CONFIDENCE VS CORRECTNESS ANALYSIS COMPLETE")
print("=" * 90)
```

```python
# Purpose: Identify and save individual overconfident incorrect predictions for all six models

print("=" * 90)
print("OBJECTIVE 5 — OVERCONFIDENT ERROR CASE ANALYSIS")
print("=" * 90)

overconfident_error_records = []

for model_name, df in probability_data.items():

    y_true = df["true_label"].to_numpy(dtype=np.int64)
    predictions = df["predicted_label"].to_numpy(dtype=np.int64)
    melanoma_probability = df["melanoma_probability"].to_numpy(
        dtype=np.float64
    )

    confidence = np.where(
        predictions == 1,
        melanoma_probability,
        1.0 - melanoma_probability
    )

    correct = predictions == y_true

    error_mask = ~correct

    for index in np.where(error_mask)[0]:

        true_label = int(y_true[index])
        predicted_label = int(predictions[index])
        probability = float(melanoma_probability[index])
        prediction_confidence = float(confidence[index])

        if true_label == 0 and predicted_label == 1:
            error_type = "False Positive"
        elif true_label == 1 and predicted_label == 0:
            error_type = "False Negative"
        else:
            error_type = "Unknown"

        overconfident_error_records.append({
            "Model": model_name,
            "sample_index": int(
                df.iloc[index]["sample_index"]
            ),
            "true_label": true_label,
            "predicted_label": predicted_label,
            "melanoma_probability": probability,
            "confidence": prediction_confidence,
            "error_type": error_type,
            "high_confidence_80": prediction_confidence >= 0.80,
            "very_high_confidence_90": prediction_confidence >= 0.90
        })

overconfident_errors_df = pd.DataFrame(
    overconfident_error_records
)

overconfident_errors_df = overconfident_errors_df.sort_values(
    by=["Model", "confidence"],
    ascending=[True, False]
).reset_index(drop=True)

all_errors_path = (
    OBJECTIVE5_DIR /
    "Six_Model_All_Incorrect_Predictions.csv"
)

overconfident_errors_df.to_csv(
    all_errors_path,
    index=False
)

print("\nTotal incorrect predictions across all models:")
print(len(overconfident_errors_df))

print("\nIncorrect predictions by model:")

incorrect_counts = (
    overconfident_errors_df
    .groupby("Model")
    .size()
    .sort_values(ascending=False)
)

print(incorrect_counts.to_string())

print("\nOverconfident errors at confidence >= 0.80:")

errors_80 = (
    overconfident_errors_df[
        overconfident_errors_df["confidence"] >= 0.80
    ]
    .groupby("Model")
    .size()
)

print(errors_80.to_string())

print("\nVery-high-confidence errors at confidence >= 0.90:")

errors_90 = (
    overconfident_errors_df[
        overconfident_errors_df["confidence"] >= 0.90
    ]
    .groupby("Model")
    .size()
)

print(errors_90.to_string())

print("\nTop 10 most overconfident errors:")

top_10_errors = (
    overconfident_errors_df
    .sort_values(
        by="confidence",
        ascending=False
    )
    .head(10)
)

print(
    top_10_errors[
        [
            "Model",
            "sample_index",
            "true_label",
            "predicted_label",
            "melanoma_probability",
            "confidence",
            "error_type"
        ]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

print("\nSaved artifact:")
print(all_errors_path)

print("\n" + "=" * 90)
print("OVERCONFIDENT ERROR CASE ANALYSIS COMPLETE")
print("=" * 90)
```

```python
# Purpose: Quantify entropy-based predictive uncertainty for all six models

import numpy as np
import pandas as pd

print("=" * 90)
print("OBJECTIVE 5 — ENTROPY-BASED UNCERTAINTY ANALYSIS")
print("=" * 90)

entropy_results = []
entropy_records = {}

MAX_BINARY_ENTROPY = np.log(2.0)

for model_name, df in probability_data.items():

    y_true = df["true_label"].to_numpy(dtype=np.int64)

    melanoma_probability = df[
        "melanoma_probability"
    ].to_numpy(dtype=np.float64)

    predictions = df[
        "predicted_label"
    ].to_numpy(dtype=np.int64)

    confidence = np.where(
        predictions == 1,
        melanoma_probability,
        1.0 - melanoma_probability
    )

    correct = predictions == y_true

    p = np.clip(
        melanoma_probability,
        1e-12,
        1.0 - 1e-12
    )

    entropy = (
        -(p * np.log(p))
        - ((1.0 - p) * np.log(1.0 - p))
    )

    normalized_entropy = (
        entropy / MAX_BINARY_ENTROPY
    )

    high_uncertainty = entropy >= 0.60
    very_high_uncertainty = entropy >= 0.65

    high_uncertainty_errors = (
        high_uncertainty & ~correct
    )

    very_high_uncertainty_errors = (
        very_high_uncertainty & ~correct
    )

    correct_entropy = entropy[correct]
    incorrect_entropy = entropy[~correct]

    entropy_results.append({
        "Model": model_name,
        "Mean Entropy": entropy.mean(),
        "Median Entropy": np.median(entropy),
        "Std Entropy": entropy.std(),
        "Mean Normalized Entropy": normalized_entropy.mean(),
        "Mean Correct Entropy": (
            correct_entropy.mean()
            if len(correct_entropy) > 0
            else np.nan
        ),
        "Mean Incorrect Entropy": (
            incorrect_entropy.mean()
            if len(incorrect_entropy) > 0
            else np.nan
        ),
        "High-Uncertainty Predictions >= 0.60": int(
            high_uncertainty.sum()
        ),
        "High-Uncertainty Errors >= 0.60": int(
            high_uncertainty_errors.sum()
        ),
        "Very-High-Uncertainty Predictions >= 0.65": int(
            very_high_uncertainty.sum()
        ),
        "Very-High-Uncertainty Errors >= 0.65": int(
            very_high_uncertainty_errors.sum()
        )
    })

    entropy_records[model_name] = pd.DataFrame({
        "sample_index": df["sample_index"].to_numpy(),
        "true_label": y_true,
        "predicted_label": predictions,
        "melanoma_probability": melanoma_probability,
        "confidence": confidence,
        "entropy": entropy,
        "normalized_entropy": normalized_entropy,
        "correct": correct
    })

entropy_summary = pd.DataFrame(
    entropy_results
).sort_values(
    by="Mean Entropy",
    ascending=True
).reset_index(drop=True)

print("\nEntropy uncertainty summary:")

print(
    entropy_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

entropy_output_path = (
    OBJECTIVE5_DIR /
    "Six_Model_Entropy_Uncertainty_Summary.csv"
)

entropy_summary.to_csv(
    entropy_output_path,
    index=False
)

print("\nSaved summary:")
print(entropy_output_path)

print("\nMaximum possible binary entropy:")
print(f"{MAX_BINARY_ENTROPY:.6f}")

print("\n" + "=" * 90)
print("ENTROPY-BASED UNCERTAINTY ANALYSIS COMPLETE")
print("=" * 90)
```

```python
# Purpose: Create a rigorous future dermatologist reader-study protocol for clinical validation

from pathlib import Path

OBJECTIVE6_DIR = Path(
    "/kaggle/working/Objective6_Clinical_Validation"
)

OBJECTIVE6_DIR.mkdir(
    parents=True,
    exist_ok=True
)

protocol_text = """
==========================================================================================
OBJECTIVE 6 — DERMATOLOGIST READER-STUDY PROTOCOL
==========================================================================================

Purpose
-------
Define a future controlled reader study for direct comparison between the proposed
melanoma classification system and qualified dermatology clinicians.

IMPORTANT STATUS
----------------
This document defines a prospective/future validation protocol only.

No dermatologist reader-study data are currently available in this project.
Therefore, no model-versus-dermatologist performance claims are made.

1. STUDY POPULATION
-------------------
The study should recruit appropriately qualified dermatology clinicians.

The reader panel should ideally include multiple independent readers with documented
clinical experience in skin-lesion assessment.

Reader characteristics should be recorded, where appropriate, including:

- Clinical qualification
- Dermatology specialization
- Years of clinical experience
- Relevant dermoscopy experience
- Relevant melanoma diagnostic experience

Reader identities should be handled using study-specific anonymized identifiers.

2. IMAGE SET
------------
The reader study should use a predefined, locked image set that is not altered
after reader evaluation begins.

Images should represent the intended clinical-use population and should contain
both melanoma and non-melanoma lesions.

The final reference diagnosis should be established independently of the
experimental model predictions and reader predictions.

3. REFERENCE STANDARD
---------------------
The strongest reference standard should use an appropriate independent clinical
reference, preferably histopathological confirmation where available and clinically
appropriate.

The reference-standard procedure should be documented before the reader study.

Cases without sufficiently reliable reference information should be handled using
a predefined exclusion or adjudication rule rather than being selectively removed
after observing model performance.

4. READER TASK
--------------
Each dermatologist should independently classify each lesion.

For a binary melanoma-versus-non-melanoma study, the primary reader decision should
be recorded as:

- Melanoma
- Non-melanoma

Where appropriate, readers may additionally provide a confidence rating.

The reader study should use standardized instructions and a consistent evaluation
interface.

5. BLINDING
-----------
Readers should be blinded to:

- Model predictions
- Predictions from other readers
- Any post-hoc model analysis
- Study outcomes not available during the intended clinical decision

The model should be evaluated without access to the dermatologist decisions.

6. PRIMARY PERFORMANCE METRICS
------------------------------
Model and dermatologist performance should be compared using the same reference
standard.

Primary metrics should include:

- Sensitivity
- Specificity
- ROC-AUC
- Precision
- F1-score
- Accuracy

For melanoma screening, sensitivity and false-negative performance should receive
particular attention because missed melanoma cases can have clinically important
consequences.

7. SECONDARY ANALYSES
---------------------
Additional analyses should include:

- Confusion matrices
- Positive predictive value
- Negative predictive value
- Calibration
- Confidence versus correctness
- Performance by lesion subtype
- Performance by relevant demographic subgroup
- Performance by image acquisition characteristics
- Inter-reader agreement

Where statistically appropriate, confidence intervals should be reported.

8. INTER-READER AGREEMENT
-------------------------
Agreement between dermatologists should be quantified using an appropriate
agreement statistic.

For binary categorical decisions, an appropriate agreement analysis may include
pairwise or multi-reader agreement measures.

The analysis should distinguish disagreement between readers from disagreement
between readers and the reference standard.

9. MODEL-VERSUS-READER COMPARISON
----------------------------------
The comparison should use identical cases and the same reference standard.

Model and reader performance should not be compared using different datasets or
different case-selection rules.

Statistical comparison methods should be selected according to the paired nature
of the reader and model predictions.

The study should report uncertainty around performance differences rather than
relying only on point estimates.

10. READER-ASSISTED EVALUATION
-----------------------------
A separate reader-assistance experiment may evaluate whether dermatologist
performance changes when model predictions or explanations are available.

This should be treated as a separate study condition.

Possible conditions include:

- Dermatologist alone
- Model alone
- Dermatologist with model assistance

If assistance is studied, the protocol should prevent information leakage between
study conditions.

11. EXPLAINABILITY ASSESSMENT
-----------------------------
If Grad-CAM, LIME, SHAP, or attention visualizations are shown to clinicians,
their effect should be evaluated separately from the primary diagnostic comparison.

The study should not assume that an explanation is clinically useful merely because
it is visually plausible.

12. SAMPLE SIZE
---------------
The reader study should determine sample size before data collection.

Sample-size planning should account for:

- Expected melanoma prevalence
- Desired confidence interval width
- Target sensitivity and specificity
- Number of readers
- Expected reader-model performance difference
- Statistical power
- Correlation caused by multiple readers evaluating the same cases

13. DATA SPLITTING
-----------------
No image used to determine model parameters, hyperparameters, or model-selection
decisions should be reused as an independent clinical validation case.

The clinical reader-study dataset should remain locked until the evaluation protocol
has been finalized.

14. REPORTING
-------------
The final study report should clearly distinguish:

- Internal validation
- External validation
- Retrospective clinical validation
- Prospective clinical validation
- Reader-study evidence
- Clinical effectiveness evidence

Performance claims should be limited to the evidence actually generated.

15. CURRENT PROJECT STATUS
--------------------------
The current project has:

- Held-out HAM10000 evaluation
- Derm7pt cross-dataset evaluation
- PAD-UFES-20 independent external validation
- PAD-UFES-20 biopsy-status analysis
- Explainability analysis
- Fairness analysis
- Uncertainty and calibration analysis

The current project does NOT have:

- Dermatologist reader predictions
- Prospective dermatologist comparison
- Prospective clinical workflow evaluation
- Patient-outcome evaluation

Therefore, the present evidence supports a research-stage computational validation
claim, not a clinical-effectiveness or dermatologist-superiority claim.

==========================================================================================
END OF DERMATOLOGIST READER-STUDY PROTOCOL
==========================================================================================
"""

protocol_path = (
    OBJECTIVE6_DIR /
    "Objective6_Dermatologist_Reader_Study_Protocol.txt"
)

protocol_path.write_text(
    protocol_text.strip(),
    encoding="utf-8"
)

print("=" * 90)
print("OBJECTIVE 6 — DERMATOLOGIST READER-STUDY PROTOCOL CREATED")
print("=" * 90)

print("\nSaved artifact:")
print(protocol_path)

print(
    f"\nFile size: "
    f"{protocol_path.stat().st_size / 1024:.2f} KB"
)

print("\nThe protocol explicitly documents:")
print("✓ Future dermatologist reader-study design")
print("✓ Independent reference standard")
print("✓ Reader blinding")
print("✓ Model-versus-reader metrics")
print("✓ Inter-reader agreement")
print("✓ Statistical comparison requirements")
print("✓ Sample-size considerations")
print("✓ Reader-assistance evaluation")
print("✓ Current evidence limitations")

print("\nNo model training or inference was performed.")
print("No existing Objective 1–5 artifacts were modified.")

print("\n" + "=" * 90)
print("OBJECTIVE 6 READER-STUDY PROTOCOL COMPLETE")
print("=" * 90)
```

```python
# Purpose: Quantify cross-dataset performance changes from HAM10000 to external datasets

from pathlib import Path
import pandas as pd

MASTER_DIR = Path(
    "/kaggle/working/Master_Project_Audit"
)

results_path = (
    MASTER_DIR /
    "Master_Six_Model_Three_Dataset_Results.csv"
)

master_results_df = pd.read_csv(results_path)

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC_AUC"
]

external_datasets = [
    "Derm7pt",
    "PAD-UFES-20"
]

generalization_rows = []

for model_name in master_results_df["Model"].unique():

    ham_row = master_results_df[
        (master_results_df["Dataset"] == "HAM10000") &
        (master_results_df["Model"] == model_name)
    ].iloc[0]

    for external_dataset in external_datasets:

        external_row = master_results_df[
            (master_results_df["Dataset"] == external_dataset) &
            (master_results_df["Model"] == model_name)
        ].iloc[0]

        row = {
            "Model": model_name,
            "External_Dataset": external_dataset
        }

        for metric in metric_columns:

            ham_value = ham_row[metric]
            external_value = external_row[metric]

            absolute_change = external_value - ham_value

            percentage_point_change = (
                absolute_change * 100
            )

            relative_change_percent = (
                (absolute_change / ham_value) * 100
                if ham_value != 0
                else None
            )

            row[f"HAM10000_{metric}"] = ham_value
            row[f"External_{metric}"] = external_value
            row[f"{metric}_Change"] = absolute_change
            row[f"{metric}_Change_PP"] = percentage_point_change
            row[f"{metric}_Relative_Change_Percent"] = relative_change_percent

        generalization_rows.append(row)

generalization_df = pd.DataFrame(
    generalization_rows
)

generalization_path = (
    MASTER_DIR /
    "Cross_Dataset_Generalization_Analysis.csv"
)

generalization_df.to_csv(
    generalization_path,
    index=False
)

print("=" * 100)
print("CROSS-DATASET GENERALIZATION ANALYSIS")
print("=" * 100)

for external_dataset in external_datasets:

    print("\n" + "-" * 100)
    print(
        f"HAM10000 → {external_dataset}"
    )
    print("-" * 100)

    display_columns = [
        "Model",
        "Accuracy_Change_PP",
        "Precision_Change_PP",
        "Recall_Change_PP",
        "F1_Change_PP",
        "ROC_AUC_Change_PP"
    ]

    display(
        generalization_df[
            generalization_df["External_Dataset"] == external_dataset
        ][display_columns].round(2)
    )

print("\n" + "-" * 100)
print("BEST AND WORST GENERALIZATION BY METRIC")
print("-" * 100)

for external_dataset in external_datasets:

    dataset_df = generalization_df[
        generalization_df["External_Dataset"] == external_dataset
    ]

    print(
        f"\nHAM10000 → {external_dataset}"
    )

    for metric in metric_columns:

        change_column = (
            f"{metric}_Change_PP"
        )

        best_row = dataset_df.loc[
            dataset_df[change_column].idxmax()
        ]

        worst_row = dataset_df.loc[
            dataset_df[change_column].idxmin()
        ]

        print(
            f"{metric}: "
            f"Best = {best_row['Model']} "
            f"({best_row[change_column]:+.2f} pp), "
            f"Worst = {worst_row['Model']} "
            f"({worst_row[change_column]:+.2f} pp)"
        )

print("\n" + "-" * 100)

print("Saved artifact:")
print(generalization_path)

print(
    f"\nFile size: "
    f"{generalization_path.stat().st_size / 1024:.2f} KB"
)

print("\nNo model training or inference was performed.")
print("No existing experimental artifacts were modified.")

print("\n" + "=" * 100)
print("CROSS-DATASET GENERALIZATION ANALYSIS COMPLETE")
print("=" * 100)
```

```python
# Purpose: Rank the six models across the three evaluated datasets using multiple performance metrics

from pathlib import Path
import pandas as pd

MASTER_DIR = Path(
    "/kaggle/working/Master_Project_Audit"
)

results_path = (
    MASTER_DIR /
    "Master_Six_Model_Three_Dataset_Results.csv"
)

master_results_df = pd.read_csv(results_path)

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC_AUC"
]

ranking_rows = []

for model_name in master_results_df["Model"].unique():

    model_df = master_results_df[
        master_results_df["Model"] == model_name
    ].copy()

    row = {
        "Model": model_name
    }

    for metric in metric_columns:
        row[f"Mean_{metric}"] = model_df[metric].mean()

    ranking_rows.append(row)

overall_ranking_df = pd.DataFrame(
    ranking_rows
)

rank_columns = []

for metric in metric_columns:

    rank_column = f"{metric}_Rank"

    overall_ranking_df[rank_column] = (
        overall_ranking_df[
            f"Mean_{metric}"
        ]
        .rank(
            ascending=False,
            method="min"
        )
    )

    rank_columns.append(rank_column)

overall_ranking_df["Mean_Metric_Rank"] = (
    overall_ranking_df[rank_columns].mean(axis=1)
)

overall_ranking_df["Overall_Rank"] = (
    overall_ranking_df["Mean_Metric_Rank"]
    .rank(
        ascending=True,
        method="min"
    )
)

overall_ranking_df = overall_ranking_df.sort_values(
    "Overall_Rank"
)

ranking_path = (
    MASTER_DIR /
    "Overall_Six_Model_Three_Dataset_Ranking.csv"
)

overall_ranking_df.to_csv(
    ranking_path,
    index=False
)

print("=" * 100)
print("OVERALL SIX-MODEL THREE-DATASET RANKING")
print("=" * 100)

display(
    overall_ranking_df[
        [
            "Overall_Rank",
            "Model",
            "Mean_Accuracy",
            "Mean_Precision",
            "Mean_Recall",
            "Mean_F1",
            "Mean_ROC_AUC",
            "Mean_Metric_Rank"
        ]
    ].round(4)
)

print("\n" + "-" * 100)

print("Metric-specific overall leaders:")

for metric in metric_columns:

    best_row = overall_ranking_df.loc[
        overall_ranking_df[
            f"Mean_{metric}"
        ].idxmax()
    ]

    print(
        f"{metric}: "
        f"{best_row['Model']} "
        f"({best_row[f'Mean_{metric}'] * 100:.2f}%)"
    )

print("\n" + "-" * 100)

overall_best = overall_ranking_df.iloc[0]

print(
    f"Overall multi-metric leader: "
    f"{overall_best['Model']}"
)

print(
    f"Mean metric rank: "
    f"{overall_best['Mean_Metric_Rank']:.2f}"
)

print("\n" + "-" * 100)

print("Saved artifact:")
print(ranking_path)

print(
    f"\nFile size: "
    f"{ranking_path.stat().st_size / 1024:.2f} KB"
)

print("\nNo model training or inference was performed.")
print("No existing experimental artifacts were modified.")

print("\n" + "=" * 100)
print("OVERALL MODEL RANKING COMPLETE")
print("=" * 100)
```

```python
# Step 15: Compare LAMNet against the five benchmark models across all datasets

from pathlib import Path
import pandas as pd

MASTER_AUDIT_DIR = Path("/kaggle/working/Master_Project_Audit")

results_path = MASTER_AUDIT_DIR / "Master_Six_Model_Three_Dataset_Results.csv"

results_df = pd.read_csv(results_path)

benchmark_params = {
    "AlexNet": 57012034,
    "VGG16": 134268738,
    "VGG19": 139578434,
    "ResNet50": 23512130,
    "EfficientNet-B0": 4010110,
    "LAMNet": 2427148
}

summary_rows = []

for model_name, model_df in results_df.groupby("Model"):
    summary_rows.append({
        "Model": model_name,
        "Parameters": benchmark_params.get(model_name),
        "Mean_Accuracy": model_df["Accuracy"].mean(),
        "Mean_Precision": model_df["Precision"].mean(),
        "Mean_Recall": model_df["Recall"].mean(),
        "Mean_F1": model_df["F1"].mean(),
        "Mean_ROC_AUC": model_df["ROC_AUC"].mean()
    })

lamnet_comparison = pd.DataFrame(summary_rows)

for metric in [
    "Mean_Accuracy",
    "Mean_Precision",
    "Mean_Recall",
    "Mean_F1",
    "Mean_ROC_AUC"
]:
    lamnet_comparison[f"{metric}_Percent"] = (
        lamnet_comparison[metric] * 100
    )

lamnet_comparison["Parameter_Millions"] = (
    lamnet_comparison["Parameters"] / 1_000_000
)

lamnet_comparison["F1_Per_Million_Params"] = (
    lamnet_comparison["Mean_F1"] /
    lamnet_comparison["Parameter_Millions"]
)

lamnet_comparison = lamnet_comparison.sort_values(
    "Mean_F1",
    ascending=False
).reset_index(drop=True)

comparison_path = (
    MASTER_AUDIT_DIR /
    "LAMNet_vs_Five_Benchmarks_Comparison.csv"
)

lamnet_comparison.to_csv(comparison_path, index=False)

print("=" * 100)
print("LAMNet VS FIVE BENCHMARK MODELS")
print("=" * 100)

display(
    lamnet_comparison[
        [
            "Model",
            "Parameter_Millions",
            "Mean_Accuracy_Percent",
            "Mean_Precision_Percent",
            "Mean_Recall_Percent",
            "Mean_F1_Percent",
            "Mean_ROC_AUC_Percent",
            "F1_Per_Million_Params"
        ]
    ]
)

print("\n" + "-" * 100)
print("LAMNet parameter count:")
print(f"{benchmark_params['LAMNet']:,}")

print("\n" + "-" * 100)
print("Best mean F1 model:")
best_f1 = lamnet_comparison.iloc[0]
print(f"{best_f1['Model']} ({best_f1['Mean_F1'] * 100:.2f}%)")

print("\n" + "-" * 100)
print("Best mean recall model:")
best_recall = lamnet_comparison.loc[
    lamnet_comparison["Mean_Recall"].idxmax()
]
print(f"{best_recall['Model']} ({best_recall['Mean_Recall'] * 100:.2f}%)")

print("\n" + "-" * 100)
print("Most parameter-efficient model by F1 per million parameters:")
best_efficiency = lamnet_comparison.loc[
    lamnet_comparison["F1_Per_Million_Params"].idxmax()
]
print(
    f"{best_efficiency['Model']} "
    f"({best_efficiency['F1_Per_Million_Params']:.4f})"
)

print("\n" + "-" * 100)
print("Saved artifact:")
print(comparison_path)
print(f"File size: {comparison_path.stat().st_size / 1024:.2f} KB")

print("\nNo model training or inference was performed.")
print("No existing experimental artifacts were modified.")

print("\n" + "=" * 100)
print("LAMNET VS BENCHMARK COMPARISON COMPLETE")
print("=" * 100)
```
