```python
# Create the Derm7pt dermoscopic dataset for external evaluation

import os
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

class Derm7ptExternalDataset(Dataset):
    def __init__(self, dataframe, image_root, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.image_root = image_root
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        relative_path = row["derm"].replace("\\", "/")
        image_path = os.path.join(
            self.image_root,
            relative_path
        )

        image = Image.open(image_path).convert("RGB")
        label = int(row["label"])

        if self.transform is not None:
            image = self.transform(image)

        return image, label


# Use the same ImageNet normalization as HAM10000 evaluation
derm7pt_eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


derm7pt_test_dataset = Derm7ptExternalDataset(
    dataframe=derm7pt_test,
    image_root=DERM7PT_IMAGE_ROOT,
    transform=derm7pt_eval_transform
)


derm7pt_test_loader = DataLoader(
    derm7pt_test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


print("=" * 60)
print("DERM7PT EXTERNAL EVALUATION DATASET")
print("=" * 60)

print("Test cases:", len(derm7pt_test_dataset))
print("Batches:", len(derm7pt_test_loader))

print("\nExpected labels:")
print("Non-Melanoma:", (derm7pt_test["label"] == 0).sum())
print("Melanoma:", (derm7pt_test["label"] == 1).sum())

print("\n" + "=" * 60)
print("DATASET READY")
print("=" * 60)
```

```python
# Load the trained HAM10000 checkpoints from their saved training checkpoints

import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model_checkpoint_paths = {
    "AlexNet": "/kaggle/input/datasets/akansha1009/5-trained-models/"
               "HAM10000_AlexNet_Balanced_Baseline_Backup/"
               "ham10000_alexnet_balanced_best.pth",

    "VGG16": "/kaggle/input/datasets/akansha1009/5-trained-models/"
             "HAM10000_VGG16_Balanced_Baseline_Backup/"
             "ham10000_vgg16_balanced_best.pth",

    "VGG19": "/kaggle/input/datasets/akansha1009/5-trained-models/"
             "HAM10000_VGG19_Balanced_Baseline_Backup/"
             "ham10000_vgg19_balanced_best.pth",

    "ResNet50": "/kaggle/input/datasets/akansha1009/5-trained-models/"
                "HAM10000_ResNet50_Balanced_Baseline_Backup/"
                "ham10000_resnet50_balanced_best.pth",

    "EfficientNet-B0": "/kaggle/input/datasets/akansha1009/5-trained-models/"
                       "HAM10000_EfficientNetB0_Balanced_Baseline_Backup/"
                       "ham10000_efficientnet_b0_balanced_best.pth"
}


model_architectures = {
    "AlexNet": alexnet,
    "VGG16": vgg16,
    "VGG19": vgg19,
    "ResNet50": resnet50,
    "EfficientNet-B0": efficientnet_b0
}


loaded_models = {}

print("=" * 60)
print("LOADING FIVE TRAINED HAM10000 MODELS")
print("=" * 60)

for name, model in model_architectures.items():

    checkpoint = torch.load(
        model_checkpoint_paths[name],
        map_location="cpu",
        weights_only=False
    )

    state_dict = checkpoint["model_state_dict"]

    cleaned_state_dict = {}

    for key, value in state_dict.items():
        if key.startswith("module."):
            key = key.replace("module.", "", 1)

        cleaned_state_dict[key] = value

    model.load_state_dict(cleaned_state_dict)

    model = model.to(device)
    model.eval()

    loaded_models[name] = model

    print(
        f"{name}: LOADED | "
        f"Best epoch: {checkpoint.get('epoch', 'N/A')} | "
        f"Best Val F1: {checkpoint.get('best_val_f1', 'N/A')}"
    )


print("\nDevice:", device)
print("Models loaded:", len(loaded_models))

print("\n" + "=" * 60)
print("ALL FIVE TRAINED MODELS LOADED SUCCESSFULLY")
print("=" * 60)
```

```python
# Evaluate all five frozen HAM10000 models on the untouched Derm7pt test set

import torch
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

print("=" * 60)
print("DERM7PT EXTERNAL EVALUATION")
print("=" * 60)

external_results = {}

criterion = torch.nn.CrossEntropyLoss()

for name, model in loaded_models.items():

    print(f"\nEvaluating {name}...")

    model.eval()

    all_labels = []
    all_predictions = []
    all_probabilities = []
    total_loss = 0.0
    total_samples = 0

    with torch.no_grad():

        for images, labels in derm7pt_test_loader:

            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            outputs = model(images)

            loss = criterion(outputs, labels)

            probabilities = torch.softmax(outputs, dim=1)[:, 1]
            predictions = (probabilities >= 0.5).long()

            batch_size = labels.size(0)

            total_loss += loss.item() * batch_size
            total_samples += batch_size

            all_labels.extend(labels.cpu().numpy())
            all_predictions.extend(predictions.cpu().numpy())
            all_probabilities.extend(probabilities.cpu().numpy())

    all_labels = np.array(all_labels)
    all_predictions = np.array(all_predictions)
    all_probabilities = np.array(all_probabilities)

    test_loss = total_loss / total_samples

    accuracy = accuracy_score(all_labels, all_predictions)
    precision = precision_score(
        all_labels,
        all_predictions,
        zero_division=0
    )
    recall = recall_score(
        all_labels,
        all_predictions,
        zero_division=0
    )
    f1 = f1_score(
        all_labels,
        all_predictions,
        zero_division=0
    )
    roc_auc = roc_auc_score(
        all_labels,
        all_probabilities
    )

    cm = confusion_matrix(
        all_labels,
        all_predictions
    )

    external_results[name] = {
        "test_loss": test_loss,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "roc_auc": roc_auc,
        "confusion_matrix": cm.tolist()
    }

    print(
        f"Loss: {test_loss:.4f} | "
        f"Accuracy: {accuracy:.4f} | "
        f"Precision: {precision:.4f} | "
        f"Recall: {recall:.4f} | "
        f"F1: {f1:.4f} | "
        f"ROC-AUC: {roc_auc:.4f}"
    )

    print("Confusion Matrix:")
    print(cm)

print("\n" + "=" * 60)
print("DERM7PT EXTERNAL EVALUATION COMPLETE")
print("=" * 60)
```

```python
# --- Restore EfficientNet-B0 class weights and training configuration ---

import torch
import torch.nn as nn
import torch.optim as optim

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

NUM_CLASSES = 2
NUM_EPOCHS = 20
LEARNING_RATE = 1e-4
WEIGHT_DECAY = 1e-5

class_counts = train_df["binary_label"].value_counts().sort_index()

class_weights = len(train_df) / (
    len(class_counts) * class_counts
)

class_weights = torch.tensor(
    class_weights.values,
    dtype=torch.float32,
    device=DEVICE
)

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

print("Device:", DEVICE)
print("Number of epochs:", NUM_EPOCHS)
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Class weights:", class_weights)
print("AMP enabled:", torch.cuda.is_available())
```

```python
# --- Restore the EfficientNet-B0 binary classification model ---

import torch
import torch.nn as nn
from torchvision.models import efficientnet_b0, EfficientNet_B0_Weights

model = efficientnet_b0(
    weights=EfficientNet_B0_Weights.DEFAULT
)

model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    NUM_CLASSES
)

model = model.to(DEVICE)

print("Model:", "EfficientNet-B0")
print("Parameters:", sum(p.numel() for p in model.parameters()))
print("Device:", DEVICE)
```

```python
# --- Test EfficientNet-B0 optimizer, AMP, forward pass, and one training batch ---

from torch.amp import autocast, GradScaler

optimizer = optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

scaler = GradScaler("cuda", enabled=torch.cuda.is_available())

model.train()

images, labels = next(iter(train_loader))

images = images.to(DEVICE, non_blocking=True)
labels = labels.to(DEVICE, non_blocking=True)

optimizer.zero_grad(set_to_none=True)

with autocast("cuda", enabled=torch.cuda.is_available()):
    outputs = model(images)
    loss = criterion(outputs, labels)

scaler.scale(loss).backward()
scaler.step(optimizer)
scaler.update()

print("Single-batch test passed.")
print("Input shape:", images.shape)
print("Output shape:", outputs.shape)
print("Loss:", float(loss))
print("Optimizer:", type(optimizer).__name__)
print("AMP:", torch.cuda.is_available())
```

```python
# --- Train EfficientNet-B0 from a clean initialization for 20 epochs ---

import time
import copy
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

model = efficientnet_b0(
    weights=EfficientNet_B0_Weights.DEFAULT
)

model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    NUM_CLASSES
)

model = model.to(DEVICE)

optimizer = optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

scaler = GradScaler("cuda", enabled=torch.cuda.is_available())

best_val_f1 = -1.0
best_epoch = 0
best_model_state = None

history = {
    "train_loss": [],
    "val_loss": [],
    "val_accuracy": [],
    "val_precision": [],
    "val_recall": [],
    "val_f1": [],
    "val_roc_auc": []
}

start_time = time.time()

for epoch in range(1, NUM_EPOCHS + 1):

    model.train()

    running_loss = 0.0
    train_samples = 0

    for images, labels in train_loader:

        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        with autocast("cuda", enabled=torch.cuda.is_available()):
            outputs = model(images)
            loss = criterion(outputs, labels)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        batch_size = labels.size(0)
        running_loss += loss.detach().item() * batch_size
        train_samples += batch_size

    train_loss = running_loss / train_samples

    model.eval()

    val_loss_total = 0.0
    val_samples = 0
    val_labels = []
    val_predictions = []
    val_probabilities = []

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            with autocast("cuda", enabled=torch.cuda.is_available()):
                outputs = model(images)
                loss = criterion(outputs, labels)

            probabilities = torch.softmax(outputs, dim=1)[:, 1]
            predictions = torch.argmax(outputs, dim=1)

            batch_size = labels.size(0)

            val_loss_total += loss.detach().item() * batch_size
            val_samples += batch_size

            val_labels.extend(labels.cpu().numpy())
            val_predictions.extend(predictions.cpu().numpy())
            val_probabilities.extend(probabilities.cpu().numpy())

    val_loss = val_loss_total / val_samples

    val_accuracy = accuracy_score(val_labels, val_predictions)
    val_precision = precision_score(
        val_labels,
        val_predictions,
        zero_division=0
    )
    val_recall = recall_score(
        val_labels,
        val_predictions,
        zero_division=0
    )
    val_f1 = f1_score(
        val_labels,
        val_predictions,
        zero_division=0
    )
    val_roc_auc = roc_auc_score(
        val_labels,
        val_probabilities
    )

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["val_accuracy"].append(val_accuracy)
    history["val_precision"].append(val_precision)
    history["val_recall"].append(val_recall)
    history["val_f1"].append(val_f1)
    history["val_roc_auc"].append(val_roc_auc)

    if val_f1 > best_val_f1:

        best_val_f1 = val_f1
        best_epoch = epoch
        best_model_state = copy.deepcopy(model.state_dict())

    elapsed = (time.time() - start_time) / 60

    print(
        f"Epoch {epoch:02d}/{NUM_EPOCHS} | "
        f"Train Loss {train_loss:.4f} | "
        f"Val Loss {val_loss:.4f} | "
        f"Val Acc {val_accuracy:.4f} | "
        f"Val F1 {val_f1:.4f} | "
        f"Val AUC {val_roc_auc:.4f} | "
        f"Time {elapsed:.2f} min"
    )

print("\nTraining completed.")
print("Best validation F1:", best_val_f1)
print("Best epoch:", best_epoch)
```

```python
# --- Evaluate the best EfficientNet-B0 checkpoint on the untouched test set ---

checkpoint = torch.load(
    EFFICIENTNET_BEST_PATH,
    map_location=DEVICE
)

model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

test_loss_total = 0.0
test_samples = 0

test_labels = []
test_predictions = []
test_probabilities = []

test_start_time = time.time()

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        with autocast("cuda", enabled=torch.cuda.is_available()):
            outputs = model(images)
            loss = criterion(outputs, labels)

        probabilities = torch.softmax(outputs, dim=1)[:, 1]
        predictions = torch.argmax(outputs, dim=1)

        batch_size = labels.size(0)

        test_loss_total += loss.detach().item() * batch_size
        test_samples += batch_size

        test_labels.extend(labels.cpu().numpy())
        test_predictions.extend(predictions.cpu().numpy())
        test_probabilities.extend(probabilities.cpu().numpy())

test_loss = test_loss_total / test_samples

test_accuracy = accuracy_score(
    test_labels,
    test_predictions
)

test_precision = precision_score(
    test_labels,
    test_predictions,
    zero_division=0
)

test_recall = recall_score(
    test_labels,
    test_predictions,
    zero_division=0
)

test_f1 = f1_score(
    test_labels,
    test_predictions,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    test_labels,
    test_probabilities
)

from sklearn.metrics import confusion_matrix

test_confusion_matrix = confusion_matrix(
    test_labels,
    test_predictions
)

test_elapsed = time.time() - test_start_time

print("EfficientNet-B0 test evaluation completed.")
print("\nTest Loss:", round(test_loss, 6))
print("Accuracy:", round(test_accuracy, 6))
print("Precision:", round(test_precision, 6))
print("Recall:", round(test_recall, 6))
print("F1 Score:", round(test_f1, 6))
print("ROC-AUC:", round(test_roc_auc, 6))

print("\nConfusion Matrix:")
print(test_confusion_matrix)

print("\nEvaluation time:",
      round(test_elapsed / 60, 2),
      "minutes")

print("Average time per image:",
      round(test_elapsed / len(test_dataset), 4),
      "seconds")
```

```python
# DERM7PT PYTORCH DATASET RESTORATION
# Purpose: Rebuild the train, validation, and test PyTorch
#          datasets from the frozen Derm7pt metadata.

from PIL import Image
from torch.utils.data import Dataset


class Derm7ptDataset(Dataset):

    def __init__(
        self,
        dataframe,
        image_root,
        transform=None
    ):

        self.dataframe = (
            dataframe.reset_index(
                drop=True
            )
        )

        self.image_root = image_root
        self.transform = transform


    def __len__(self):

        return len(
            self.dataframe
        )


    def __getitem__(
        self,
        index
    ):

        row = self.dataframe.iloc[index]

        image_path = row["derm_path"]

        relative_path = (
            image_path
            .replace(
                "/content/drive/MyDrive/"
                "Melanoma Skin Lesion Analysis Project/"
                "data/Derm7pt/raw/release_v0/",
                ""
            )
        )

        full_path = (
            self.image_root
            .replace(
                "/images",
                ""
            )
            + "/"
            + relative_path
        )


        image = Image.open(
            full_path
        ).convert("RGB")


        label = (
            1
            if row["binary_label"] == "Melanoma"
            else 0
        )


        if self.transform is not None:

            image = self.transform(
                image
            )


        return (
            image,
            label
        )


train_df = df[
    df["split"] == "train"
].copy()

val_df = df[
    df["split"] == "validation"
].copy()

test_df = df[
    df["split"] == "test"
].copy()


train_dataset = Derm7ptDataset(
    train_df,
    IMAGE_ROOT
)

val_dataset = Derm7ptDataset(
    val_df,
    IMAGE_ROOT
)

test_dataset = Derm7ptDataset(
    test_df,
    IMAGE_ROOT
)


sample_image, sample_label = (
    train_dataset[0]
)


print("DERM7PT PYTORCH DATASET RESTORATION")

print("\nTraining samples:")
print(len(train_dataset))

print("\nValidation samples:")
print(len(val_dataset))

print("\nTest samples:")
print(len(test_dataset))

print("\nSample image type:")
print(type(sample_image))

print("\nSample image size:")
print(sample_image.size)

print("\nSample label:")
print(sample_label)

print("✅ DERM7PT PYTORCH DATASETS RESTORED")
```

```python
# DERM7PT EFFICIENTNET-B0 BASELINE TRAINING
# Purpose: Train the clean EfficientNet-B0 baseline for 20
#          epochs with validation monitoring and checkpointing.

import os
import time
import numpy as np
import torch
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


EFFICIENTNET_BEST_PATH = (
    "/kaggle/working/"
    "derm7pt_efficientnet_b0_best.pth"
)

EFFICIENTNET_HISTORY_PATH = (
    "/kaggle/working/"
    "derm7pt_efficientnet_b0_training_history.pt"
)


efficientnet_history = {
    "train_loss": [],
    "val_loss": [],
    "val_accuracy": [],
    "val_precision": [],
    "val_recall": [],
    "val_f1": [],
    "val_roc_auc": []
}


best_efficientnet_f1 = -1.0
best_efficientnet_epoch = 0


def evaluate_efficientnet(
    model,
    loader
):

    model.eval()

    total_loss = 0.0

    all_labels = []
    all_predictions = []
    all_probabilities = []


    with torch.no_grad():

        for images, labels in loader:

            images = images.to(
                DEVICE,
                non_blocking=True
            )

            labels = labels.to(
                DEVICE,
                non_blocking=True
            )


            with torch.amp.autocast(
                device_type="cuda"
            ):

                outputs = model(
                    images
                )

                loss = efficientnet_criterion(
                    outputs,
                    labels
                )


            probabilities = torch.softmax(
                outputs.float(),
                dim=1
            )[:, 1]


            predictions = torch.argmax(
                outputs,
                dim=1
            )


            total_loss += (
                loss.item()
                * images.size(0)
            )


            all_labels.extend(
                labels.cpu().numpy()
            )

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_probabilities.extend(
                probabilities.cpu().numpy()
            )


    all_labels = np.array(
        all_labels
    )

    all_predictions = np.array(
        all_predictions
    )

    all_probabilities = np.array(
        all_probabilities
    )


    average_loss = (
        total_loss
        / len(loader.dataset)
    )


    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    precision = precision_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    recall = recall_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    f1 = f1_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        all_labels,
        all_probabilities
    )


    return (
        average_loss,
        accuracy,
        precision,
        recall,
        f1,
        roc_auc
    )


print("DERM7PT EFFICIENTNET-B0 BASELINE TRAINING")

print(
    f"\nTraining for {EPOCHS} epochs..."
)


training_start = time.time()


for epoch in range(
    1,
    EPOCHS + 1
):

    epoch_start = time.time()

    efficientnet_model.train()

    running_loss = 0.0


    for images, labels in train_loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        labels = labels.to(
            DEVICE,
            non_blocking=True
        )


        efficientnet_optimizer.zero_grad(
            set_to_none=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = efficientnet_model(
                images
            )

            loss = efficientnet_criterion(
                outputs,
                labels
            )


        efficientnet_scaler.scale(
            loss
        ).backward()


        efficientnet_scaler.step(
            efficientnet_optimizer
        )

        efficientnet_scaler.update()


        running_loss += (
            loss.item()
            * images.size(0)
        )


    train_loss = (
        running_loss
        / len(train_loader.dataset)
    )


    (
        val_loss,
        val_accuracy,
        val_precision,
        val_recall,
        val_f1,
        val_roc_auc
    ) = evaluate_efficientnet(
        efficientnet_model,
        val_loader
    )


    efficientnet_history[
        "train_loss"
    ].append(
        train_loss
    )

    efficientnet_history[
        "val_loss"
    ].append(
        val_loss
    )

    efficientnet_history[
        "val_accuracy"
    ].append(
        val_accuracy
    )

    efficientnet_history[
        "val_precision"
    ].append(
        val_precision
    )

    efficientnet_history[
        "val_recall"
    ].append(
        val_recall
    )

    efficientnet_history[
        "val_f1"
    ].append(
        val_f1
    )

    efficientnet_history[
        "val_roc_auc"
    ].append(
        val_roc_auc
    )


    epoch_time = (
        time.time()
        - epoch_start
    )


    is_best = (
        val_f1
        > best_efficientnet_f1
    )


    if is_best:

        best_efficientnet_f1 = val_f1
        best_efficientnet_epoch = epoch

        torch.save(
            efficientnet_model.state_dict(),
            EFFICIENTNET_BEST_PATH
        )


    marker = (
        " ⭐ BEST"
        if is_best
        else ""
    )


    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"Val Precision: {val_precision:.4f} | "
        f"Val Recall: {val_recall:.4f} | "
        f"Val F1: {val_f1:.4f} | "
        f"Val AUC: {val_roc_auc:.4f} | "
        f"Time: {epoch_time / 60:.2f} min"
        f"{marker}"
    )


total_training_time = (
    time.time()
    - training_start
)


torch.save(
    efficientnet_history,
    EFFICIENTNET_HISTORY_PATH
)


print(
    "DERM7PT EFFICIENTNET-B0 BASELINE TRAINING COMPLETED"
)

print("\nBest validation F1:")
print(
    f"{best_efficientnet_f1:.6f}"
)

print("\nBest epoch:")
print(
    best_efficientnet_epoch
)

print("\nBest model:")
print(
    EFFICIENTNET_BEST_PATH
)

print("\nTraining history:")
print(
    EFFICIENTNET_HISTORY_PATH
)

print("\nTotal training time:")
print(
    f"{total_training_time / 60:.2f} minutes"
)

```

```python
# DERM7PT EFFICIENTNET-B0 FINAL TEST SET EVALUATION
# Purpose: Evaluate the saved best-validation checkpoint on the
#          untouched Derm7pt test set and calculate final metrics.

import time
import numpy as np
import torch
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)


best_efficientnet_model.eval()


all_labels = []
all_predictions = []
all_probabilities = []

total_test_loss = 0.0


if torch.cuda.is_available():
    torch.cuda.synchronize()

evaluation_start = time.time()


with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        labels = labels.to(
            DEVICE,
            non_blocking=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = best_efficientnet_model(
                images
            )

            loss = efficientnet_criterion(
                outputs,
                labels
            )


        probabilities = torch.softmax(
            outputs.float(),
            dim=1
        )[:, 1]


        predictions = torch.argmax(
            outputs,
            dim=1
        )


        total_test_loss += (
            loss.item()
            * images.size(0)
        )


        all_labels.extend(
            labels.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_probabilities.extend(
            probabilities.cpu().numpy()
        )


if torch.cuda.is_available():
    torch.cuda.synchronize()


evaluation_time = (
    time.time()
    - evaluation_start
)


all_labels = np.array(
    all_labels
)

all_predictions = np.array(
    all_predictions
)

all_probabilities = np.array(
    all_probabilities
)


test_loss = (
    total_test_loss
    / len(test_loader.dataset)
)


test_accuracy = accuracy_score(
    all_labels,
    all_predictions
)

test_precision = precision_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_recall = recall_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_f1 = f1_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    all_labels,
    all_probabilities
)


cm = confusion_matrix(
    all_labels,
    all_predictions
)


average_time_per_image = (
    evaluation_time
    / len(test_loader.dataset)
)


print("DERM7PT EFFICIENTNET-B0 FINAL TEST SET EVALUATION")

print("\nTest samples:")
print(len(test_loader.dataset))

print("\nTest loss:")
print(f"{test_loss:.6f}")

print("\nTest Accuracy:")
print(f"{test_accuracy:.6f}")

print("\nTest Precision:")
print(f"{test_precision:.6f}")

print("\nTest Recall / Sensitivity:")
print(f"{test_recall:.6f}")

print("\nTest F1:")
print(f"{test_f1:.6f}")

print("\nTest ROC-AUC:")
print(f"{test_roc_auc:.6f}")

print("\nConfusion Matrix:")
print(cm)

print("\nTotal evaluation time:")
print(f"{evaluation_time / 60:.2f} minutes")

print("\nAverage time per test image:")
print(f"{average_time_per_image:.4f} seconds")

print("✅ EFFICIENTNET-B0 TEST EVALUATION COMPLETED")
```

```python
# DERM7PT KAGGLE IMAGE PATH VALIDATION - ABSOLUTE PATH FIX
# Purpose: Convert the frozen Google Drive image paths into
#          Kaggle-relative paths and verify every image reference.

import os
import pandas as pd


DATASET_ROOT = (
    "/kaggle/input/datasets/akansha1009/"
    "derm7pt-frozen-dataset"
)

FROZEN_CSV = os.path.join(
    DATASET_ROOT,
    "Derm7pt_split_metadata.csv"
)

IMAGE_ROOT = os.path.join(
    DATASET_ROOT,
    "Derm7pt_Kaggle_Ready",
    "images"
)


metadata = pd.read_csv(
    FROZEN_CSV
)


def convert_to_kaggle_path(original_path):

    path = str(
        original_path
    ).replace("\\", "/")

    marker = "/images/"

    if marker in path:

        relative_path = path.split(
            marker,
            1
        )[1]

    elif path.startswith("images/"):

        relative_path = path[len("images/"):]

    else:

        relative_path = path

    return os.path.join(
        IMAGE_ROOT,
        relative_path
    )


clinic_missing = []
derm_missing = []


for _, row in metadata.iterrows():

    clinic_full_path = convert_to_kaggle_path(
        row["clinic_path"]
    )

    derm_full_path = convert_to_kaggle_path(
        row["derm_path"]
    )


    if not os.path.exists(
        clinic_full_path
    ):

        clinic_missing.append(
            {
                "case_num": row["case_num"],
                "original_path": row["clinic_path"],
                "kaggle_path": clinic_full_path
            }
        )


    if not os.path.exists(
        derm_full_path
    ):

        derm_missing.append(
            {
                "case_num": row["case_num"],
                "original_path": row["derm_path"],
                "kaggle_path": derm_full_path
            }
        )


print("DERM7PT FINAL KAGGLE IMAGE PATH VALIDATION")

print("\nFrozen records:")
print(len(metadata))

print("\nClinical references:")
print(len(metadata))

print("Missing clinical images:")
print(len(clinic_missing))

print("\nDermoscopic references:")
print(len(metadata))

print("Missing dermoscopic images:")
print(len(derm_missing))


print("\nExample path conversion:")

example = metadata.iloc[0]

print("\nOriginal clinical path:")
print(example["clinic_path"])

print("\nConverted Kaggle clinical path:")
print(
    convert_to_kaggle_path(
        example["clinic_path"]
    )
)

print("\nOriginal dermoscopic path:")
print(example["derm_path"])

print("\nConverted Kaggle dermoscopic path:")
print(
    convert_to_kaggle_path(
        example["derm_path"]
    )
)



if (
    len(clinic_missing) == 0
    and len(derm_missing) == 0
):

    print(
        "✅ ALL 2,022 FROZEN IMAGE REFERENCES RESOLVE IN KAGGLE"
    )

else:

    print(
        "❌ SOME IMAGE REFERENCES STILL DO NOT RESOLVE"
    )

    if clinic_missing:
        print("\nFirst missing clinical reference:")
        print(clinic_missing[0])

    if derm_missing:
        print("\nFirst missing dermoscopic reference:")
        print(derm_missing[0])
```

```python
# DERM7PT PYTORCH CLASSIFICATION DATASET
# Purpose: Create a PyTorch Dataset using the frozen Derm7pt
#          metadata and verify image/label loading.

import os
import torch
from torch.utils.data import Dataset
from PIL import Image


class Derm7ptDataset(Dataset):

    def __init__(
        self,
        dataframe,
        image_root,
        transform=None
    ):

        self.dataframe = dataframe.reset_index(
            drop=True
        )

        self.image_root = image_root
        self.transform = transform


    def _get_image_path(self, path):

        path = str(path).replace("\\", "/")

        if "/images/" in path:
            relative_path = path.split(
                "/images/",
                1
            )[1]

        elif path.startswith("images/"):
            relative_path = path[len("images/"):]

        else:
            relative_path = path

        return os.path.join(
            self.image_root,
            relative_path
        )


    def __len__(self):

        return len(self.dataframe)


    def __getitem__(self, index):

        row = self.dataframe.iloc[index]

        image_path = self._get_image_path(
            row["derm_path"]
        )

        image = Image.open(
            image_path
        ).convert("RGB")

        label = 1 if row["binary_label"] == "Melanoma" else 0

        if self.transform is not None:
            image = self.transform(image)

        label = torch.tensor(
            label,
            dtype=torch.long
        )

        return image, label


train_df = df[
    df["split"] == "train"
].copy()

val_df = df[
    df["split"] == "validation"
].copy()

test_df = df[
    df["split"] == "test"
].copy()


train_dataset = Derm7ptDataset(
    train_df,
    IMAGE_ROOT
)

val_dataset = Derm7ptDataset(
    val_df,
    IMAGE_ROOT
)

test_dataset = Derm7ptDataset(
    test_df,
    IMAGE_ROOT
)


sample_image, sample_label = train_dataset[0]


print("DERM7PT PYTORCH DATASET")

print("\nTraining samples:")
print(len(train_dataset))

print("\nValidation samples:")
print(len(val_dataset))

print("\nTest samples:")
print(len(test_dataset))

print("\nSample image type:")
print(type(sample_image))

print("\nSample image size:")
print(sample_image.size)

print("\nSample label:")
print(sample_label.item())

print("\nSample label meaning:")
print(
    "Melanoma"
    if sample_label.item() == 1
    else "Non-Melanoma"
)

print("✅ DERM7PT PYTORCH DATASET READY")
```

```python
# DERM7PT IMAGE TRANSFORMS
# Purpose: Define training augmentation and deterministic
#          validation/test preprocessing for classification.

import torchvision.transforms as transforms


IMAGE_SIZE = 224


train_transform = transforms.Compose([
    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),
    transforms.RandomHorizontalFlip(
        p=0.5
    ),
    transforms.RandomVerticalFlip(
        p=0.2
    ),
    transforms.RandomRotation(
        degrees=15
    ),
    transforms.ColorJitter(
        brightness=0.15,
        contrast=0.15,
        saturation=0.10,
        hue=0.02
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


eval_transform = transforms.Compose([
    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


print("DERM7PT IMAGE TRANSFORMS")

print("\nImage size:")
print(
    f"{IMAGE_SIZE} × {IMAGE_SIZE}"
)

print("\nTraining:")
print("Resize")
print("Random horizontal flip")
print("Random vertical flip")
print("Random rotation")
print("Color jitter")
print("Normalization")

print("\nValidation/Test:")
print("Resize")
print("Normalization")

print("✅ DERM7PT IMAGE TRANSFORMS READY")
```

```python
# DERM7PT CLASS-WEIGHTED CLASSIFICATION LOSS
# Purpose: Calculate training-set class weights and define the
#          weighted CrossEntropyLoss for melanoma classification.
import torch
import torch.nn as nn


class_counts = (
    train_df["binary_label"]
    .value_counts()
)


non_melanoma_count = int(
    class_counts.get(
        "Non-Melanoma",
        0
    )
)

melanoma_count = int(
    class_counts.get(
        "Melanoma",
        0
    )
)


total_samples = (
    non_melanoma_count
    + melanoma_count
)


class_weight_non_melanoma = (
    total_samples
    / (2 * non_melanoma_count)
)

class_weight_melanoma = (
    total_samples
    / (2 * melanoma_count)
)


class_weights = torch.tensor(
    [
        class_weight_non_melanoma,
        class_weight_melanoma
    ],
    dtype=torch.float32,
    device=DEVICE
)


criterion = nn.CrossEntropyLoss(
    weight=class_weights
)


print("DERM7PT CLASS-WEIGHTED CLASSIFICATION LOSS")

print("\nTraining samples:")
print(total_samples)

print("\nNon-Melanoma:")
print(non_melanoma_count)

print("\nMelanoma:")
print(melanoma_count)

print("\nClass weights:")
print(
    class_weights.detach()
    .cpu()
    .numpy()
)

print("\nLoss:")
print("Weighted CrossEntropyLoss")

print("\nLoss device:")
print(class_weights.device)

print("✅ CLASS-WEIGHTED LOSS READY")
```

```python
# DERM7PT RESNET-18 BASELINE TRAINING
# Purpose: Train the clean Derm7pt ResNet-18 baseline for 20
#          epochs with validation metrics and best-checkpoint saving.

import os
import time
import json
import numpy as np
import torch
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


history = {
    "train_loss": [],
    "val_loss": [],
    "val_accuracy": [],
    "val_precision": [],
    "val_recall": [],
    "val_f1": [],
    "val_roc_auc": []
}


best_f1 = -1.0
best_epoch = 0


def evaluate_model(model, loader):

    model.eval()

    total_loss = 0.0
    all_labels = []
    all_predictions = []
    all_probabilities = []

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(
                DEVICE,
                non_blocking=True
            )

            labels = labels.to(
                DEVICE,
                non_blocking=True
            )

            with torch.amp.autocast(
                device_type="cuda"
            ):

                outputs = model(
                    images
                )

                loss = criterion(
                    outputs,
                    labels
                )

            probabilities = torch.softmax(
                outputs.float(),
                dim=1
            )[:, 1]

            predictions = torch.argmax(
                outputs,
                dim=1
            )

            total_loss += (
                loss.item()
                * images.size(0)
            )

            all_labels.extend(
                labels.cpu().numpy()
            )

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_probabilities.extend(
                probabilities.cpu().numpy()
            )


    all_labels = np.array(
        all_labels
    )

    all_predictions = np.array(
        all_predictions
    )

    all_probabilities = np.array(
        all_probabilities
    )


    average_loss = (
        total_loss
        / len(loader.dataset)
    )


    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    precision = precision_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    recall = recall_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    f1 = f1_score(
        all_labels,
        all_predictions,
        zero_division=0
    )


    try:

        roc_auc = roc_auc_score(
            all_labels,
            all_probabilities
        )

    except ValueError:

        roc_auc = float("nan")


    return (
        average_loss,
        accuracy,
        precision,
        recall,
        f1,
        roc_auc
    )


print("DERM7PT RESNET-18 BASELINE TRAINING")

print(
    f"\nTraining for {EPOCHS} epochs..."
)

training_start = time.time()


for epoch in range(
    1,
    EPOCHS + 1
):

    epoch_start = time.time()

    model.train()

    running_loss = 0.0


    for images, labels in train_loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        labels = labels.to(
            DEVICE,
            non_blocking=True
        )


        optimizer.zero_grad(
            set_to_none=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = model(
                images
            )

            loss = criterion(
                outputs,
                labels
            )


        scaler.scale(
            loss
        ).backward()


        scaler.step(
            optimizer
        )

        scaler.update()


        running_loss += (
            loss.item()
            * images.size(0)
        )


    train_loss = (
        running_loss
        / len(train_loader.dataset)
    )


    (
        val_loss,
        val_accuracy,
        val_precision,
        val_recall,
        val_f1,
        val_roc_auc
    ) = evaluate_model(
        model,
        val_loader
    )


    history["train_loss"].append(
        train_loss
    )

    history["val_loss"].append(
        val_loss
    )

    history["val_accuracy"].append(
        val_accuracy
    )

    history["val_precision"].append(
        val_precision
    )

    history["val_recall"].append(
        val_recall
    )

    history["val_f1"].append(
        val_f1
    )

    history["val_roc_auc"].append(
        val_roc_auc
    )


    epoch_time = (
        time.time()
        - epoch_start
    )


    is_best = (
        val_f1 > best_f1
    )


    if is_best:

        best_f1 = val_f1
        best_epoch = epoch

        torch.save(
            model.state_dict(),
            BEST_MODEL_PATH
        )


    marker = (
        " ⭐ BEST"
        if is_best
        else ""
    )


    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"Val Precision: {val_precision:.4f} | "
        f"Val Recall: {val_recall:.4f} | "
        f"Val F1: {val_f1:.4f} | "
        f"Val AUC: {val_roc_auc:.4f} | "
        f"Time: {epoch_time / 60:.2f} min"
        f"{marker}"
    )


total_training_time = (
    time.time()
    - training_start
)


torch.save(
    history,
    TRAINING_HISTORY_PATH
)


print("DERM7PT RESNET-18 BASELINE TRAINING COMPLETED")
print("=" * 60)

print("\nBest validation F1:")
print(f"{best_f1:.6f}")

print("\nBest epoch:")
print(best_epoch)

print("\nBest model:")
print(BEST_MODEL_PATH)

print("\nTraining history:")
print(TRAINING_HISTORY_PATH)

print("\nTotal training time:")
print(
    f"{total_training_time / 60:.2f} minutes"
)
```

```python
# DERM7PT RESNET-18 FINAL TEST SET EVALUATION
# Purpose: Evaluate the saved best checkpoint on the untouched
#          Derm7pt test set and calculate research metrics.

import time
import numpy as np
import torch
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)


best_model.eval()


all_labels = []
all_predictions = []
all_probabilities = []

total_test_loss = 0.0


if torch.cuda.is_available():
    torch.cuda.synchronize()

evaluation_start = time.time()


with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        labels = labels.to(
            DEVICE,
            non_blocking=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = best_model(
                images
            )

            loss = criterion(
                outputs,
                labels
            )


        probabilities = torch.softmax(
            outputs.float(),
            dim=1
        )[:, 1]


        predictions = torch.argmax(
            outputs,
            dim=1
        )


        total_test_loss += (
            loss.item()
            * images.size(0)
        )


        all_labels.extend(
            labels.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_probabilities.extend(
            probabilities.cpu().numpy()
        )


if torch.cuda.is_available():
    torch.cuda.synchronize()


evaluation_time = (
    time.time()
    - evaluation_start
)


all_labels = np.array(
    all_labels
)

all_predictions = np.array(
    all_predictions
)

all_probabilities = np.array(
    all_probabilities
)


test_loss = (
    total_test_loss
    / len(test_loader.dataset)
)


test_accuracy = accuracy_score(
    all_labels,
    all_predictions
)

test_precision = precision_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_recall = recall_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_f1 = f1_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    all_labels,
    all_probabilities
)


cm = confusion_matrix(
    all_labels,
    all_predictions
)


average_time_per_image = (
    evaluation_time
    / len(test_loader.dataset)
)


print("DERM7PT RESNET-18 FINAL TEST SET EVALUATION")

print("\nTest samples:")
print(len(test_loader.dataset))

print("\nTest loss:")
print(f"{test_loss:.6f}")

print("\nTest Accuracy:")
print(f"{test_accuracy:.6f}")

print("\nTest Precision:")
print(f"{test_precision:.6f}")

print("\nTest Recall / Sensitivity:")
print(f"{test_recall:.6f}")

print("\nTest F1:")
print(f"{test_f1:.6f}")

print("\nTest ROC-AUC:")
print(f"{test_roc_auc:.6f}")

print("\nConfusion Matrix:")
print(cm)

print("\nTotal evaluation time:")
print(f"{evaluation_time / 60:.2f} minutes")

print("\nAverage time per test image:")
print(f"{average_time_per_image:.4f} seconds")

print("✅ DERM7PT TEST EVALUATION COMPLETED")
```
